```python
"""
SDNET2018 — LOOPED CrackViT + GIPA — STACK LOOP (6 blocks x 2) — Single Cell
=============================================================================
Nanbeige-style depth sharing: the WHOLE stack of 6 transformer blocks is
applied twice with the same weights.

    stem -> cls_pos -> [block_0 ... block_5] x 2 -> head
                                          |
                                          +-- after pass 1 -> head  (exit1)

    unique blocks        : 6   (same parameters as the untied 6-block model)
    block applications   : 12 at the final exit, 6 at exit1

What is added on top of the untied model
  * PassEmbedding    : one learned vector per pass (zero-init), so the shared
                       blocks know whether they are on pass 1 or 2.
  * Per-pass gates   : GIPA's lam_c / lam_a / lam_i / gamma / kappa get one row
                       per pass; q/k/v, the tensor head and W stay shared.
  * Exit after pass 1: same head, auxiliary loss weight AUX_WEIGHT.  Gives the
                       "does pass 2 help?" number and confidence early exit.
  * Optional input injection (INPUT_INJECTION, off by default).

BATCH_SIZE = 16.  For a fair comparison rerun the untied 6-block model at
batch 16 as well -- batch size changes results on its own.

Paste the whole file into ONE Kaggle cell and run.
"""

# ═══════════════════════════════════════════════════════════════
# 1.  IMPORTS & SETUP
# ═══════════════════════════════════════════════════════════════
import os, math, random, warnings
from functools import partial

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models, callbacks
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import mixed_precision

from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    classification_report, confusion_matrix, ConfusionMatrixDisplay,
    precision_recall_curve, roc_auc_score, f1_score,
)

warnings.filterwarnings("ignore")
print(f"TensorFlow {tf.__version__}  |  Keras {keras.__version__}")
print(f"GPUs: {tf.config.list_physical_devices('GPU')}")

mixed_precision.set_global_policy("mixed_float16")
print(f"Compute dtype: {mixed_precision.global_policy().compute_dtype}")

SEED = 42
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)


# ═══════════════════════════════════════════════════════════════
# 2.  CONFIGURATION
# ═══════════════════════════════════════════════════════════════
DATA_ROOT = "/kaggle/input/datasets/aniruddhsharma/structural-defects-network-concrete-crack-images"
WORK_DIR  = "/kaggle/working/split"

ATTENTION  = "GIPA_FULL"   # MHSA | GIPA_S | GIPA_G | GIPA_I | GIPA_GI | GIPA_FULL
STEM       = "conv"        # "conv" (recommended) | "patch"

# ---- looping ------------------------------------------------------------
LOOP_MODE = "stack"        # fixed for this cell
N_PRELUDE = 0              # untied blocks before the loop
N_CORE    = 6              # shared blocks inside the loop
N_CODA    = 0              # untied blocks after the loop
N_PASSES  = 2              # times the core stack is applied
PASS_EMBED      = True     # learned 256-d vector per pass, zero-init
PER_PASS_GATES  = True     # GIPA gates (lam_c, lam_a, lam_i, gamma, kappa)
                           # untied per pass; attention weights stay shared
INPUT_INJECTION = False    # re-add the loop input on passes >= 2 (zero-init gate)
AUX_WEIGHT      = 0.3      # loss weight on the early exit(s); final exit = 1.0

IMG_SIZE   = 224
BATCH_SIZE = 16            # reduced from 32: 2 passes ~double activation memory
EPOCHS     = 60
LR_INIT    = 3e-4

EMBED_DIM  = 256
NUM_HEADS  = 8
MLP_RATIO  = 2.0
DROPOUT    = 0.1
PATCH_SIZE = 16            # only used when STEM == "patch"
COHERENCE_REG = 0.0        # optional orientation-smoothness penalty (try 1e-3)

EXIT_NAMES = [f"exit{p + 1}" for p in range(N_PASSES - 1)] + ["final"]
# block applications needed to reach each exit (the coda runs at every exit)
EXIT_COST = {name: N_PRELUDE + (p + 1) * N_CORE + N_CODA
             for p, name in enumerate(EXIT_NAMES)}
N_UNIQUE_BLOCKS = N_PRELUDE + N_CORE + N_CODA

RUN_NAME  = (f"loopvit_{LOOP_MODE}_{N_PRELUDE}-{N_CORE}x{N_PASSES}-{N_CODA}"
             f"_{ATTENTION.lower()}_{STEM}_bs{BATCH_SIZE}")
CKPT_PATH = f"best_{RUN_NAME}.weights.h5"
print(f"\n>>> RUN: {RUN_NAME}")
print(f"    unique blocks: {N_UNIQUE_BLOCKS}   block applications per exit: {EXIT_COST}")


# ═══════════════════════════════════════════════════════════════
# 3.  DATA PREPARATION
# ═══════════════════════════════════════════════════════════════
records = []
for surface in ["Decks", "Pavements", "Walls"]:
    for label in ["Cracked", "Non-cracked"]:
        folder = os.path.join(DATA_ROOT, surface, label)
        for fname in os.listdir(folder):
            records.append(dict(path=os.path.join(folder, fname),
                                label=label, surface=surface,
                                fname=f"{surface}_{fname}"))

df = pd.DataFrame(records)
print(f"\nTotal images: {len(df):,}")
print(df.groupby(["surface", "label"]).size().unstack(fill_value=0))

df["strat_key"] = df["surface"] + "_" + df["label"]
train_df, temp_df = train_test_split(df, test_size=0.30, random_state=SEED,
                                     stratify=df["strat_key"])
val_df, test_df = train_test_split(temp_df, test_size=0.50, random_state=SEED,
                                   stratify=temp_df["strat_key"])
print(f"Split -> train {len(train_df):,} | val {len(val_df):,} | test {len(test_df):,}")

for split_name, split_df in [("train", train_df), ("val", val_df), ("test", test_df)]:
    for label in ["Cracked", "Non-cracked"]:
        dst_dir = os.path.join(WORK_DIR, split_name, label)
        os.makedirs(dst_dir, exist_ok=True)
        for _, row in split_df[split_df["label"] == label].iterrows():
            dst = os.path.join(dst_dir, row["fname"])
            if not os.path.exists(dst):
                os.symlink(row["path"], dst)
print("Symlinked directory tree OK\n")


# ═══════════════════════════════════════════════════════════════
# 4.  DATA GENERATORS
# ═══════════════════════════════════════════════════════════════
train_datagen = ImageDataGenerator(
    rescale=1.0/255, rotation_range=20,
    width_shift_range=0.15, height_shift_range=0.15,
    horizontal_flip=True, vertical_flip=True,
    zoom_range=0.15, brightness_range=(0.8, 1.2),
    fill_mode="nearest",
)
val_test_datagen = ImageDataGenerator(rescale=1.0/255)

train_gen = train_datagen.flow_from_directory(
    os.path.join(WORK_DIR, "train"), target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE, class_mode="binary", shuffle=True, seed=SEED)
val_gen = val_test_datagen.flow_from_directory(
    os.path.join(WORK_DIR, "val"), target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE, class_mode="binary", shuffle=False)
test_gen = val_test_datagen.flow_from_directory(
    os.path.join(WORK_DIR, "test"), target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE, class_mode="binary", shuffle=False)

idx_to_class = {v: k for k, v in train_gen.class_indices.items()}
print(f"Class indices: {train_gen.class_indices}")

fig, axes = plt.subplots(2, 5, figsize=(14, 6))
bx, by = next(train_gen)
for i, ax in enumerate(axes.flat):
    ax.imshow(bx[i]); ax.set_title(idx_to_class[int(by[i])], fontsize=9); ax.axis("off")
plt.suptitle("Training Batch (augmented)", fontsize=13); plt.tight_layout(); plt.show()


# ═══════════════════════════════════════════════════════════════
# 5.  CLASS WEIGHTS  +  MULTI-EXIT DATA WRAPPER
# ═══════════════════════════════════════════════════════════════
train_labels = train_df["label"].map(train_gen.class_indices).values
w = compute_class_weight("balanced", classes=np.array([0, 1]), y=train_labels)
class_weights = {0: w[0], 1: w[1]}
print(f"Class weights: {idx_to_class[0]}(0)={w[0]:.3f}  {idx_to_class[1]}(1)={w[1]:.3f}")

# Keras 3 refuses `class_weight=` on a multi-output model, so the same class
# weights are applied as per-sample weights, identically on every exit.
_BaseSeq = getattr(keras.utils, "PyDataset", keras.utils.Sequence)


class MultiExitData(_BaseSeq):
    """Wraps a flow_from_directory iterator: one label copy per exit."""

    def __init__(self, it, cw=None):
        super().__init__()
        self.it, self.cw = it, cw

    def __len__(self):
        return len(self.it)

    def __getitem__(self, i):
        x, y = self.it[i]
        y = y.astype("float32")
        ys = {n: y for n in EXIT_NAMES}
        if self.cw is None:
            return x, ys
        sw = np.where(y == 1, self.cw[1], self.cw[0]).astype("float32")
        return x, ys, {n: sw for n in EXIT_NAMES}

    def on_epoch_end(self):
        self.it.on_epoch_end()


train_seq = MultiExitData(train_gen, class_weights)
val_seq   = MultiExitData(val_gen)
test_seq  = MultiExitData(test_gen)


# ═══════════════════════════════════════════════════════════════
# 6.  TOKENISERS
# ═══════════════════════════════════════════════════════════════

class PatchEmbedding(layers.Layer):
    """Plain ViT tokeniser: non-overlapping patches -> linear projection."""

    def __init__(self, patch_size, embed_dim, **kwargs):
        super().__init__(**kwargs)
        self.patch_size, self.embed_dim = patch_size, embed_dim
        self.proj = layers.Conv2D(embed_dim, patch_size, strides=patch_size,
                                  padding="valid")
        self.norm = layers.LayerNormalization(epsilon=1e-6)

    def call(self, x, training=False):
        x = self.proj(x)
        x = tf.reshape(x, [tf.shape(x)[0], -1, self.embed_dim])
        return self.norm(x)

    def get_config(self):
        c = super().get_config()
        c.update(patch_size=self.patch_size, embed_dim=self.embed_dim)
        return c


class ConvStem(layers.Layer):
    """Convolutional tokeniser: /16 downsample, 224 -> 14x14 tokens.
    Stride-2 convs, not max-pooling: MaxPool keeps the brightest pixel and
    would erase a thin dark crack."""

    def __init__(self, embed_dim, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim = embed_dim

        def blk(c, stride):
            return [layers.Conv2D(c, 3, strides=stride, padding="same",
                                  use_bias=False),
                    layers.BatchNormalization(),
                    layers.Activation("relu")]

        self.stem = keras.Sequential(
            blk(32, 1) + blk(32, 2) +          # 224 -> 112
            blk(64, 1) + blk(64, 2) +          # 112 -> 56
            blk(128, 1) + blk(128, 2) +        # 56  -> 28
            blk(embed_dim, 2),                 # 28  -> 14
            name="conv_stem")
        self.norm = layers.LayerNormalization(epsilon=1e-6)

    def call(self, x, training=False):
        x = self.stem(x, training=training)
        x = tf.reshape(x, [tf.shape(x)[0], -1, self.embed_dim])
        return self.norm(x)

    def get_config(self):
        c = super().get_config(); c.update(embed_dim=self.embed_dim); return c


class CLSPositionEmbedding(layers.Layer):
    """Prepend a learnable [CLS] token, add learnable position embeddings.
    Applied ONCE, before the loop."""

    def __init__(self, num_patches, embed_dim, dropout=0.1, **kwargs):
        super().__init__(**kwargs)
        self.num_patches, self.embed_dim = num_patches, embed_dim
        self.drop = layers.Dropout(dropout)

    def build(self, input_shape):
        self.cls_token = self.add_weight(
            name="cls_token", shape=(1, 1, self.embed_dim),
            initializer="zeros", trainable=True)
        self.pos_embed = self.add_weight(
            name="pos_embed", shape=(1, self.num_patches + 1, self.embed_dim),
            initializer=keras.initializers.TruncatedNormal(stddev=0.02),
            trainable=True)

    def call(self, x, training=False):
        B = tf.shape(x)[0]
        cls = tf.broadcast_to(tf.cast(self.cls_token, x.dtype), [B, 1, self.embed_dim])
        x = tf.concat([cls, x], axis=1) + tf.cast(self.pos_embed, x.dtype)
        return self.drop(x, training=training)

    def get_config(self):
        c = super().get_config()
        c.update(num_patches=self.num_patches, embed_dim=self.embed_dim)
        return c


# ═══════════════════════════════════════════════════════════════
# 7.  LOOP HELPERS  (pass embedding, input injection)
# ═══════════════════════════════════════════════════════════════

class PassEmbedding(layers.Layer):
    """Tells the shared blocks which pass they are on.  One learned vector per
    pass, added to every token at the start of that pass.  Zero-init, so the
    model starts as a plain weight-tied loop and learns to use it."""

    def __init__(self, num_passes, embed_dim, **kwargs):
        super().__init__(**kwargs)
        self.num_passes, self.embed_dim = num_passes, embed_dim

    def build(self, input_shape):
        self.table = self.add_weight(name="pass_table",
                                     shape=(self.num_passes, self.embed_dim),
                                     initializer="zeros", trainable=True)

    def call(self, x, pass_idx=0):
        e = tf.cast(self.table, x.dtype)[pass_idx]
        return x + e[None, None, :]

    def get_config(self):
        c = super().get_config()
        c.update(num_passes=self.num_passes, embed_dim=self.embed_dim)
        return c


class InputInjection(layers.Layer):
    """h <- h + g * x0 on passes >= 2 (Geiping et al. style, additive form).
    Per-channel gate, zero-init: starts identical to plain feedback."""

    def __init__(self, embed_dim, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim = embed_dim

    def build(self, input_shape):
        self.gate = self.add_weight(name="inject_gate", shape=(self.embed_dim,),
                                    initializer="zeros", trainable=True)

    def call(self, h, x0):
        return h + tf.cast(self.gate, h.dtype)[None, None, :] * x0

    def get_config(self):
        c = super().get_config(); c.update(embed_dim=self.embed_dim); return c


# ═══════════════════════════════════════════════════════════════
# 8.  ATTENTION — MHSA BASELINE
# ═══════════════════════════════════════════════════════════════

class MultiHeadSelfAttention(layers.Layer):
    """Standard scaled dot-product MHSA.  `num_passes` / `pass_idx` are
    accepted for interface compatibility and ignored."""

    def __init__(self, embed_dim, num_heads, dropout=0.0, num_passes=1, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim, self.num_heads = embed_dim, num_heads
        self.num_passes = num_passes
        self.head_dim = embed_dim // num_heads
        self.scale = self.head_dim ** -0.5
        self.qkv = layers.Dense(embed_dim * 3, use_bias=False)
        self.proj = layers.Dense(embed_dim)
        self.attn_drop = layers.Dropout(dropout)
        self.proj_drop = layers.Dropout(dropout)
        self.last_attn_weights = None

    def call(self, x, pass_idx=0, training=False):
        B, N = tf.shape(x)[0], tf.shape(x)[1]
        qkv = tf.reshape(self.qkv(x), [B, N, 3, self.num_heads, self.head_dim])
        qkv = tf.transpose(qkv, [2, 0, 3, 1, 4])
        q, k, v = qkv[0], qkv[1], qkv[2]

        attn = tf.nn.softmax(
            tf.cast(tf.matmul(q, k, transpose_b=True), tf.float32) * self.scale, -1)
        self.last_attn_weights = attn
        attn = self.attn_drop(tf.cast(attn, v.dtype), training=training)

        out = tf.transpose(tf.matmul(attn, v), [0, 2, 1, 3])
        out = tf.reshape(out, [B, N, self.embed_dim])
        return self.proj_drop(self.proj(out), training=training)

    def get_config(self):
        c = super().get_config()
        c.update(embed_dim=self.embed_dim, num_heads=self.num_heads,
                 num_passes=self.num_passes)
        return c


# ═══════════════════════════════════════════════════════════════
# 9.  ATTENTION — GIPA  (Geometry-Informed Predictive Attention)
# ═══════════════════════════════════════════════════════════════
#
#   logit_ij = (q_i·k_j)/sqrt(d)    content      [standard MHSA]
#            + lam_G * G_ij         GEOMETRY     orientation coherence +
#                                                displacement alignment,
#                                                anisotropy-gated
#            + lam_I * P_ij         INFORMATION  directional predictive info
#            +         S_ij         STRUCTURE    anisotropic continuity prior
#
# LOOPED VERSION: q/k/v, the structure-tensor head and the predictor W are
# shared across passes.  The scalar gates (lam_c, lam_a, lam_i, gamma, kappa)
# have one row per pass when num_passes > 1, so the model can learn a
# schedule, e.g. strong structure prior on pass 1, more geometry on pass 2.

GATE_INIT = 0.01


class GIPA(layers.Layer):
    """Geometry-Informed Predictive Attention.  Expects N = 1 + g*g."""

    def __init__(self, embed_dim, num_heads, dropout=0.0,
                 use_geometry=True, use_info=True, use_structure=True,
                 coherence_reg=0.0, num_passes=1, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim, self.num_heads = embed_dim, num_heads
        self.head_dim = embed_dim // num_heads
        self.scale = self.head_dim ** -0.5
        self.use_geometry, self.use_info = use_geometry, use_info
        self.use_structure = use_structure
        self.coherence_reg = coherence_reg
        self.num_passes = num_passes

        self.qkv = layers.Dense(embed_dim * 3, use_bias=False)
        self.proj = layers.Dense(embed_dim)
        self.attn_drop = layers.Dropout(dropout)
        self.proj_drop = layers.Dropout(dropout)

        if use_geometry or use_structure:
            self.tensor_head = layers.Dense(
                3, name="structure_tensor", dtype="float32",
                kernel_initializer=keras.initializers.TruncatedNormal(stddev=0.02),
                bias_initializer=lambda shape, dtype=None: tf.constant(
                    [0.5, 0.0, 0.5], dtype=dtype))

        self.last_attn_weights = None
        self.last_orientation = None
        self.last_anisotropy = None

    def build(self, input_shape):
        N = int(input_shape[1]); P = N - 1
        g = int(round(math.sqrt(P)))
        if g * g != P:
            raise ValueError(f"GIPA expects N = 1 + g*g; got N={N}.")
        self.grid, self.num_patches = g, P

        rr, cc = np.meshgrid(np.arange(g), np.arange(g), indexing="ij")
        pos = np.stack([rr.ravel(), cc.ravel()], -1).astype("float32")
        delta = pos[None, :, :] - pos[:, None, :]              # (P,P,2) = p_j - p_i
        d2 = (delta ** 2).sum(-1)
        nrm = np.sqrt(d2) + 1e-8
        dx, dy = delta[..., 0] / nrm, delta[..., 1] / nrm
        disp = np.stack([dx ** 2 - dy ** 2, 2.0 * dx * dy], -1)
        disp[d2 == 0] = 0.0

        self._disp = disp.astype("float32")
        self._dist2n = (d2 / (d2.max() + 1e-8)).astype("float32")
        self._adj = (d2 == 1.0).astype("float32")

        C = keras.initializers.Constant
        G, H = self.num_passes, self.num_heads      # gate table: (passes, heads)
        if self.use_geometry:
            self.lam_c = self.add_weight(name="lam_c", shape=(G, H),
                                         initializer=C(GATE_INIT), trainable=True)
            self.lam_a = self.add_weight(name="lam_a", shape=(G, H),
                                         initializer=C(GATE_INIT), trainable=True)
        if self.use_info:
            eye = np.stack([np.eye(self.head_dim, dtype="float32")
                            for _ in range(self.num_heads)])
            self.W = self.add_weight(
                name="pred_W",
                shape=(self.num_heads, self.head_dim, self.head_dim),
                initializer=lambda shape, dtype=None: tf.constant(eye, tf.float32),
                trainable=True)
            self.lam_i = self.add_weight(name="lam_i", shape=(G, H),
                                         initializer=C(GATE_INIT), trainable=True)
        if self.use_structure:
            self.gamma = self.add_weight(name="gamma", shape=(G, H),
                                         initializer=C(-6.0), trainable=True)
            self.kappa = self.add_weight(name="kappa", shape=(G, H),
                                         initializer="zeros", trainable=True)
        super().build(input_shape)

    def _row(self, var, pass_idx):
        """Gate row for this pass, shaped (1, H, 1, 1), float32."""
        g = pass_idx if self.num_passes > 1 else 0
        return tf.cast(var, tf.float32)[g][None, :, None, None]

    def _orientation(self, x32):
        """u: (B,P,2) unit double-angle (cos2t, sin2t);  alpha: (B,P) in [0,1).
        alpha ~ 0 = isotropic (rough concrete);  ~1 = oriented ridge (crack)."""
        t = tf.cast(self.tensor_head(x32[:, 1:]), tf.float32)
        a, b, c = t[..., 0], t[..., 1], t[..., 2]
        u_raw = tf.stack([a - c, 2.0 * b], -1)
        n = tf.norm(u_raw, axis=-1)
        u = u_raw / (n[..., None] + 1e-6)
        alpha = n / (n + tf.nn.softplus(a + c) + 1e-6)
        return u, alpha

    def _info_bias(self, z):
        """Directional predictive information (negative log residual energy
        of z_j ~ W z_i), row-normalised and floored.  z: (B,H,N,dh)."""
        Wz = tf.einsum('bhnd,hde->bhne', z, tf.cast(self.W, tf.float32))
        zz = tf.reduce_sum(z * z, -1)
        ww = tf.reduce_sum(Wz * Wz, -1)
        cross = tf.matmul(Wz, z, transpose_b=True)
        e = tf.maximum(ww[..., :, None] + zz[..., None, :] - 2.0 * cross, 0.0)
        e = e / (tf.reduce_mean(e, axis=-1, keepdims=True) + 1e-6)
        return -tf.math.log(e + 0.1)

    @staticmethod
    def _pad_cls(bias):
        return tf.pad(bias, [[0, 0], [0, 0], [1, 0], [1, 0]])

    def _coherence_loss(self, u, alpha):
        adj = tf.convert_to_tensor(self._adj)[None]
        c = tf.einsum('bpk,bqk->bpq', u, u)
        wgt = alpha[:, :, None] * alpha[:, None, :]
        num = tf.reduce_sum(adj * wgt * (1.0 - c), [1, 2])
        den = tf.reduce_sum(adj * wgt, [1, 2]) + 1e-6
        return tf.reduce_mean(num / den)

    def call(self, x, pass_idx=0, training=False):
        B, N = tf.shape(x)[0], tf.shape(x)[1]
        qkv = tf.reshape(self.qkv(x), [B, N, 3, self.num_heads, self.head_dim])
        qkv = tf.transpose(qkv, [2, 0, 3, 1, 4])
        q, k, v = qkv[0], qkv[1], qkv[2]

        logits = tf.cast(tf.matmul(q, k, transpose_b=True), tf.float32) * self.scale
        x32 = tf.cast(x, tf.float32)
        u = alpha = align = gate = None

        if self.use_geometry or self.use_structure:
            u, alpha = self._orientation(x32)
            self.last_orientation, self.last_anisotropy = u, alpha
            disp = tf.convert_to_tensor(self._disp)
            align = 0.5 * (tf.einsum('pqk,bpk->bpq', disp, u)
                           + tf.einsum('pqk,bqk->bpq', disp, u))
            gate = tf.sqrt(alpha[:, :, None] * alpha[:, None, :] + 1e-8)

        if self.use_geometry:
            coh = tf.einsum('bpk,bqk->bpq', u, u)
            G = (self._row(self.lam_c, pass_idx) * coh[:, None]
                 + self._row(self.lam_a, pass_idx) * align[:, None]) * gate[:, None]
            logits += self._pad_cls(G)

        if self.use_structure:
            g_h = tf.nn.softplus(self._row(self.gamma, pass_idx))
            k_h = tf.sigmoid(self._row(self.kappa, pass_idx))
            relax = 1.0 - k_h * gate[:, None] * tf.square(align[:, None])
            S = -g_h * tf.convert_to_tensor(self._dist2n)[None, None] * relax
            logits += self._pad_cls(S)

        if self.use_info:
            logits += self._row(self.lam_i, pass_idx) * self._info_bias(
                tf.cast(v, tf.float32))

        attn = tf.nn.softmax(logits, axis=-1)
        self.last_attn_weights = attn
        attn = self.attn_drop(tf.cast(attn, v.dtype), training=training)

        out = tf.transpose(tf.matmul(attn, v), [0, 2, 1, 3])
        out = tf.reshape(out, [B, N, self.embed_dim])
        out = self.proj_drop(self.proj(out), training=training)

        if self.coherence_reg > 0.0 and u is not None:
            self.add_loss(self.coherence_reg * self._coherence_loss(u, alpha))
        return out

    def get_config(self):
        c = super().get_config()
        c.update(embed_dim=self.embed_dim, num_heads=self.num_heads,
                 use_geometry=self.use_geometry, use_info=self.use_info,
                 use_structure=self.use_structure,
                 coherence_reg=self.coherence_reg, num_passes=self.num_passes)
        return c


ATTENTION_REGISTRY = {
    "MHSA":      MultiHeadSelfAttention,
    "GIPA_S":    partial(GIPA, use_geometry=False, use_info=False, use_structure=True),
    "GIPA_G":    partial(GIPA, use_geometry=True,  use_info=False, use_structure=False),
    "GIPA_I":    partial(GIPA, use_geometry=False, use_info=True,  use_structure=False),
    "GIPA_GI":   partial(GIPA, use_geometry=True,  use_info=True,  use_structure=False),
    "GIPA_FULL": partial(GIPA, use_geometry=True,  use_info=True,  use_structure=True,
                         coherence_reg=COHERENCE_REG),
}


# ═══════════════════════════════════════════════════════════════
# 10.  TRANSFORMER BLOCK, HEAD & LOOPED MODEL
# ═══════════════════════════════════════════════════════════════

class TransformerBlock(layers.Layer):
    """Pre-LayerNorm Transformer block; forwards `pass_idx` to the attention."""

    def __init__(self, embed_dim, num_heads, mlp_ratio=2.0, dropout=0.1,
                 attention_cls=None, num_passes=1, **kwargs):
        super().__init__(**kwargs)
        AttnCls = attention_cls or MultiHeadSelfAttention
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.attn = AttnCls(embed_dim, num_heads, dropout=dropout,
                            num_passes=num_passes)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.mlp = keras.Sequential([
            layers.Dense(int(embed_dim * mlp_ratio)),
            layers.Activation("gelu"), layers.Dropout(dropout),
            layers.Dense(embed_dim), layers.Dropout(dropout),
        ])

    def call(self, x, pass_idx=0, training=False):
        x = x + self.attn(self.norm1(x), pass_idx=pass_idx, training=training)
        x = x + self.mlp(self.norm2(x), training=training)
        return x


class ClassifierHead(layers.Layer):
    """final_norm -> [CLS] -> fc -> gelu -> dropout -> sigmoid.
    ONE head shared by every exit, so exits are directly comparable."""

    def __init__(self, embed_dim, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim = embed_dim
        self.norm = layers.LayerNormalization(epsilon=1e-6, name="final_norm")
        self.fc = layers.Dense(embed_dim, name="head_fc")
        self.act = layers.Activation("gelu")
        self.drop = layers.Dropout(0.3)
        self.out = layers.Dense(1, activation="sigmoid", dtype="float32", name="pred")

    def call(self, x, training=False):
        x = self.norm(x)[:, 0]
        x = self.drop(self.act(self.fc(x)), training=training)
        return self.out(x)

    def get_config(self):
        c = super().get_config(); c.update(embed_dim=self.embed_dim); return c


def build_looped_crack_vit(attention_cls, stem="conv",
                           n_prelude=N_PRELUDE, n_core=N_CORE, n_coda=N_CODA,
                           n_passes=N_PASSES, input_shape=(IMG_SIZE, IMG_SIZE, 3),
                           patch_size=PATCH_SIZE, embed_dim=EMBED_DIM,
                           num_heads=NUM_HEADS, mlp_ratio=MLP_RATIO,
                           dropout=DROPOUT):
    """
        stem -> cls_pos -> prelude -> [ core ] x n_passes -> coda -> head
                                            |                  ^
                                            +-- after every pass: coda -> head
                                                (early exits, shared weights)
    """
    def make_block(name, passes):
        return TransformerBlock(embed_dim, num_heads, mlp_ratio, dropout,
                                attention_cls=attention_cls,
                                num_passes=passes, name=name)

    if stem == "conv":
        tokeniser = ConvStem(embed_dim, name="tokeniser")
        grid = input_shape[0] // 16
    else:
        tokeniser = PatchEmbedding(patch_size, embed_dim, name="tokeniser")
        grid = input_shape[0] // patch_size
    cls_pos = CLSPositionEmbedding(grid * grid, embed_dim, dropout, name="cls_pos")

    core_passes = n_passes if PER_PASS_GATES else 1
    prelude = [make_block(f"prelude_{i}", 1) for i in range(n_prelude)]
    core    = [make_block(f"core_{i}", core_passes) for i in range(n_core)]
    coda    = [make_block(f"coda_{i}", 1) for i in range(n_coda)]
    pass_emb = PassEmbedding(n_passes, embed_dim, name="pass_embed") if PASS_EMBED else None
    inject   = InputInjection(embed_dim, name="input_injection") if INPUT_INJECTION else None
    head     = ClassifierHead(embed_dim, name="head")

    inputs = layers.Input(shape=input_shape)
    x = cls_pos(tokeniser(inputs))
    for b in prelude:
        x = b(x, pass_idx=0)

    x0, h, outputs = x, x, {}
    for p in range(n_passes):
        if inject is not None and p > 0:
            h = inject(h, x0)
        if pass_emb is not None:
            h = pass_emb(h, pass_idx=p)
        for b in core:
            h = b(h, pass_idx=p)
        z = h
        for b in coda:
            z = b(z, pass_idx=0)
        outputs[EXIT_NAMES[p]] = head(z)

    model = models.Model(inputs, outputs,
                         name=f"LoopedCrackViT_{ATTENTION}_{LOOP_MODE}")
    parts = dict(tokeniser=tokeniser, cls_pos=cls_pos, prelude=prelude,
                 core=core, coda=coda, pass_emb=pass_emb, inject=inject, head=head)
    return model, parts


model, PARTS = build_looped_crack_vit(ATTENTION_REGISTRY[ATTENTION], stem=STEM)
model.summary()
print(f"\nTotal params: {model.count_params():,}")
print(f"Unique transformer blocks: {N_UNIQUE_BLOCKS}  |  "
      f"block applications: exit1={EXIT_COST[EXIT_NAMES[0]]}, final={EXIT_COST['final']}")


# ═══════════════════════════════════════════════════════════════
# 11.  COMPILE & CALLBACKS
# ═══════════════════════════════════════════════════════════════
def make_metrics():
    return [keras.metrics.BinaryAccuracy(name="acc"),
            keras.metrics.AUC(name="auc"),
            keras.metrics.Precision(name="prec"),
            keras.metrics.Recall(name="rec")]


model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=LR_INIT),
    loss={n: keras.losses.BinaryCrossentropy(label_smoothing=0.05) for n in EXIT_NAMES},
    loss_weights={n: (1.0 if n == "final" else AUX_WEIGHT) for n in EXIT_NAMES},
    metrics={n: make_metrics() for n in EXIT_NAMES},
)

MONITOR = "val_final_auc"
cb_list = [
    callbacks.EarlyStopping(monitor=MONITOR, patience=10, mode="max",
                            restore_best_weights=True, verbose=1),
    callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=4,
                                min_lr=1e-6, verbose=1),
    callbacks.ModelCheckpoint(CKPT_PATH, monitor=MONITOR, mode="max",
                              save_best_only=True, save_weights_only=True,
                              verbose=1),
]


# ═══════════════════════════════════════════════════════════════
# 12.  TRAIN
# ═══════════════════════════════════════════════════════════════
history = model.fit(train_seq, validation_data=val_seq, epochs=EPOCHS,
                    callbacks=cb_list, verbose=1)
H = history.history


# ═══════════════════════════════════════════════════════════════
# 13.  TRAINING CURVES  (final exit solid, early exit dashed)
# ═══════════════════════════════════════════════════════════════
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
axes[0].plot(H["loss"], label="train (total)")
axes[0].plot(H["val_loss"], label="val (total)")
axes[0].set_title("Loss", fontsize=13)
for ax, m, lab in [(axes[1], "acc", "Accuracy"), (axes[2], "auc", "AUC")]:
    for n in EXIT_NAMES:
        ls = "-" if n == "final" else "--"
        if f"{n}_{m}" in H:
            ax.plot(H[f"{n}_{m}"], ls, label=f"train {n}")
            ax.plot(H[f"val_{n}_{m}"], ls, label=f"val {n}")
    ax.set_title(lab, fontsize=13)
for ax in axes:
    ax.set_xlabel("Epoch"); ax.legend(fontsize=8); ax.grid(True, alpha=0.3)
plt.suptitle(RUN_NAME, fontsize=13)
plt.tight_layout(); plt.savefig(f"curves_{RUN_NAME}.png", dpi=150); plt.show()


# ═══════════════════════════════════════════════════════════════
# 14.  TEST EVALUATION  (every exit)
# ═══════════════════════════════════════════════════════════════
model.load_weights(CKPT_PATH)
test_gen.reset()
raw = model.predict(test_seq, verbose=1)
probs = {n: np.asarray(raw[n]).ravel() for n in EXIT_NAMES}
y_true = test_gen.classes
y_prob = probs["final"]
y_pred = (y_prob >= 0.5).astype(int)
target_names = [idx_to_class[i] for i in range(len(idx_to_class))]

print("\n" + "=" * 60)
print(f"  {RUN_NAME} — FINAL EXIT TEST METRICS")
print("=" * 60)
print(classification_report(y_true, y_pred, target_names=target_names, digits=4))

exit_rows = []
for n in EXIT_NAMES:
    pr = probs[n]; pd_ = (pr >= 0.5).astype(int)
    exit_rows.append(dict(exit=n, block_apps=EXIT_COST[n],
                          acc=float((y_true == pd_).mean()),
                          auc=float(roc_auc_score(y_true, pr)),
                          f1_macro=float(f1_score(y_true, pd_, average="macro"))))
exit_df = pd.DataFrame(exit_rows)
print("PER-EXIT (does the second pass help?)")
print(exit_df.to_string(index=False))

overall = dict(run=RUN_NAME, loop_mode=LOOP_MODE, n_prelude=N_PRELUDE,
               n_core=N_CORE, n_coda=N_CODA, n_passes=N_PASSES,
               acc=exit_rows[-1]["acc"], auc=exit_rows[-1]["auc"],
               f1_macro=exit_rows[-1]["f1_macro"])
for r in exit_rows[:-1]:
    overall.update({f"{r['exit']}_acc": r["acc"], f"{r['exit']}_auc": r["auc"],
                    f"{r['exit']}_f1_macro": r["f1_macro"]})
print(f"\nFINAL  ROC-AUC: {overall['auc']:.4f}   F1-macro: {overall['f1_macro']:.4f}")

fig, ax = plt.subplots(figsize=(6, 5))
ConfusionMatrixDisplay(confusion_matrix(y_true, y_pred),
                       display_labels=target_names).plot(ax=ax, cmap="Blues",
                                                         values_format="d")
ax.set_title(f"{RUN_NAME}\nTest Confusion Matrix (final exit)", fontsize=10)
plt.tight_layout(); plt.savefig(f"cm_{RUN_NAME}.png", dpi=150); plt.show()


# ═══════════════════════════════════════════════════════════════
# 15.  PER-SUBSET BREAKDOWN  (Decks / Pavements / Walls)
# ═══════════════════════════════════════════════════════════════
surfaces = np.array([os.path.basename(f).split("_")[0] for f in test_gen.filenames])

print("\n" + "=" * 60)
print("  PER-SUBSET TEST METRICS  (final exit; early-exit AUC alongside)")
print("=" * 60)
rows = []
for subset in ["Decks", "Pavements", "Walls"]:
    mask = surfaces == subset
    if mask.sum() == 0:
        continue
    yt, yp, ypr = y_true[mask], y_pred[mask], y_prob[mask]
    two = len(np.unique(yt)) > 1
    r = dict(Surface=subset, N=int(mask.sum()),
             Accuracy=float((yt == yp).mean()),
             AUC=roc_auc_score(yt, ypr) if two else np.nan,
             F1_macro=f1_score(yt, yp, average="macro"))
    for n in EXIT_NAMES[:-1]:
        r[f"AUC_{n}"] = roc_auc_score(yt, probs[n][mask]) if two else np.nan
    rows.append(r)
    print(f"\n--- {subset} ({mask.sum():,} images) ---")
    print(classification_report(yt, yp, target_names=target_names,
                                digits=4, zero_division=0))
subset_df = pd.DataFrame(rows)
print(subset_df.to_string(index=False))


# ═══════════════════════════════════════════════════════════════
# 16.  PRECISION-RECALL CURVE + THRESHOLD TUNING  (final exit)
# ═══════════════════════════════════════════════════════════════
cracked_idx = train_gen.class_indices["Cracked"]
y_true_c = (y_true == cracked_idx).astype(int)
y_prob_c = y_prob if cracked_idx == 1 else 1.0 - y_prob

prec, rec, thr = precision_recall_curve(y_true_c, y_prob_c)
f1s = 2 * prec[:-1] * rec[:-1] / (prec[:-1] + rec[:-1] + 1e-8)
bi = int(np.argmax(f1s))
print(f"\nOptimal threshold (Cracked-positive): {thr[bi]:.3f}")
print(f"  -> F1 {f1s[bi]:.4f} | Prec {prec[bi]:.4f} | Rec {rec[bi]:.4f}")

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(rec, prec, lw=2)
ax.scatter(rec[bi], prec[bi], s=120, c="red", zorder=5,
           label=f"Best F1={f1s[bi]:.3f} @ t={thr[bi]:.3f}")
ax.set_xlabel("Recall (Cracked)"); ax.set_ylabel("Precision (Cracked)")
ax.set_title(f"{RUN_NAME}\nPR Curve, Cracked class (final exit)", fontsize=10)
ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(f"pr_{RUN_NAME}.png", dpi=150); plt.show()


# ═══════════════════════════════════════════════════════════════
# 17.  CONFIDENCE-BASED EARLY EXIT  (no extra training)
# ═══════════════════════════════════════════════════════════════
# If exit 1 is confident (max(p, 1-p) >= tau) stop there, else take the
# final exit.  Label smoothing 0.05 caps confidence near 0.975, so useful
# tau values sit below that.
p1, pf = probs[EXIT_NAMES[0]], probs["final"]
conf1 = np.maximum(p1, 1.0 - p1)
c1, cF = EXIT_COST[EXIT_NAMES[0]], EXIT_COST["final"]
ee_rows = []
for tau in [None, 0.97, 0.95, 0.925, 0.90, 0.85, 0.80, 0.0]:
    stop = np.zeros_like(conf1, bool) if tau is None else conf1 >= tau
    pm = np.where(stop, p1, pf); pdm = (pm >= 0.5).astype(int)
    frac = float(stop.mean())
    ee_rows.append(dict(
        tau="final only" if tau is None else ("exit1 only" if tau == 0.0 else tau),
        exited_early=frac,
        avg_block_apps=frac * c1 + (1 - frac) * cF,
        acc=float((y_true == pdm).mean()),
        auc=float(roc_auc_score(y_true, pm)),
        f1_macro=float(f1_score(y_true, pdm, average="macro"))))
ee_df = pd.DataFrame(ee_rows)
print("\n" + "=" * 60)
print("  EARLY EXIT: accuracy vs average compute")
print("=" * 60)
print(ee_df.to_string(index=False))

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.plot(ee_df["avg_block_apps"], ee_df["acc"], "o-")
for _, r in ee_df.iterrows():
    ax.annotate(str(r["tau"]), (r["avg_block_apps"], r["acc"]),
                fontsize=8, xytext=(4, 4), textcoords="offset points")
ax.set_xlabel("Average block applications per image")
ax.set_ylabel("Test accuracy")
ax.set_title(f"{RUN_NAME}\nEarly-exit trade-off", fontsize=10)
ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(f"early_exit_{RUN_NAME}.png", dpi=150); plt.show()


# ═══════════════════════════════════════════════════════════════
# 18.  TRACED FORWARD PASS  (records EVERY block application)
# ═══════════════════════════════════════════════════════════════
# Shared layers overwrite last_attn_weights / last_orientation on every call,
# so the functional model only remembers the last pass.  This replays the
# final-exit path eagerly and snapshots each application as it happens.

def trace_forward(img):
    P = PARTS
    x = P["cls_pos"](P["tokeniser"](img, training=False), training=False)
    rec = []

    def snap(blk, tag):
        a = blk.attn
        rec.append(dict(tag=tag, attn=a.last_attn_weights,
                        u=getattr(a, "last_orientation", None),
                        alpha=getattr(a, "last_anisotropy", None),
                        is_gipa=isinstance(a, GIPA)))

    for b in P["prelude"]:
        x = b(x, pass_idx=0, training=False); snap(b, f"{b.name}")
    x0, h = x, x
    for p in range(N_PASSES):
        if P["inject"] is not None and p > 0:
            h = P["inject"](h, x0)
        if P["pass_emb"] is not None:
            h = P["pass_emb"](h, pass_idx=p)
        for b in P["core"]:
            h = b(h, pass_idx=p, training=False); snap(b, f"{b.name}/p{p + 1}")
    z = h
    for b in P["coda"]:
        z = b(z, pass_idx=0, training=False); snap(b, f"{b.name}")
    prob = P["head"](z, training=False)
    return rec, prob


# ═══════════════════════════════════════════════════════════════
# 19.  ATTENTION ROLLOUT  (all block applications on the final path)
# ═══════════════════════════════════════════════════════════════

def attention_rollout(recs, head_fusion="mean"):
    maps = []
    for r in recs:
        a = r["attn"]
        a = tf.reduce_mean(a, 1) if head_fusion == "mean" else tf.reduce_max(a, 1)
        maps.append(tf.cast(a, tf.float32))
    N = maps[0].shape[-1]
    roll = tf.eye(N, batch_shape=[tf.shape(maps[0])[0]])
    for a in maps:
        roll = tf.matmul(roll, 0.5 * a + 0.5 * tf.eye(N))
    cls_attn = roll[0, 0, 1:].numpy()
    g = int(np.sqrt(len(cls_attn)))
    hm = cls_attn.reshape(g, g)
    return (hm - hm.min()) / (hm.max() - hm.min() + 1e-8)


test_gen.reset()
bx, by = test_gen[0]
n_show = min(5, len(bx))
fig, axes = plt.subplots(2, n_show, figsize=(18, 7), squeeze=False)
for i in range(n_show):
    img = bx[i:i + 1]
    recs, prob = trace_forward(img)
    p = float(prob.numpy()[0, 0])
    hm = attention_rollout(recs)
    hm_up = tf.image.resize(hm[..., None], (IMG_SIZE, IMG_SIZE),
                            method="bicubic").numpy().squeeze()
    axes[0, i].imshow(bx[i]); axes[0, i].axis("off")
    axes[0, i].set_title(f"True: {idx_to_class[int(by[i])]}\n"
                         f"Pred: {idx_to_class[int(p >= 0.5)]} ({p:.2f})", fontsize=9)
    axes[1, i].imshow(bx[i]); axes[1, i].imshow(hm_up, cmap="jet", alpha=0.4)
    axes[1, i].set_title(f"Rollout over {len(recs)} applications", fontsize=9)
    axes[1, i].axis("off")
plt.suptitle(f"{RUN_NAME} — where the model attends", fontsize=13)
plt.tight_layout(); plt.savefig(f"rollout_{RUN_NAME}.png", dpi=150); plt.show()


# ═══════════════════════════════════════════════════════════════
# 20.  GIPA-ONLY DIAGNOSTICS  (per-pass gates, geometry per pass)
# ═══════════════════════════════════════════════════════════════
gipa_blocks = [b for b in PARTS["prelude"] + PARTS["core"] + PARTS["coda"]
               if isinstance(b.attn, GIPA)]

if gipa_blocks:
    sp = lambda v: np.log1p(np.exp(v))
    sg = lambda v: 1 / (1 + np.exp(-v))

    # ---- learned gate values, one row per (block, pass) ------------------
    grep = []
    for b in gipa_blocks:
        a = b.attn
        for g in range(a.num_passes):
            r = {"block": b.name, "pass": g + 1 if a.num_passes > 1 else "shared"}
            if a.use_geometry:
                r["lam_c"] = float(np.mean(np.abs(a.lam_c.numpy()[g])))
                r["lam_a"] = float(np.mean(np.abs(a.lam_a.numpy()[g])))
            if a.use_info:
                r["lam_i"] = float(np.mean(np.abs(a.lam_i.numpy()[g])))
            if a.use_structure:
                r["gamma_sp"] = float(np.mean(sp(a.gamma.numpy()[g])))
                r["kappa_sig"] = float(np.mean(sg(a.kappa.numpy()[g])))
            grep.append(r)
    print("\n" + "=" * 60)
    print("  LEARNED GATE VALUES PER PASS  (init 0.01; growth => term is used)")
    print("=" * 60)
    print(pd.DataFrame(grep).to_string(index=False))
    if PARTS["pass_emb"] is not None:
        tbl = PARTS["pass_emb"].table.numpy()
        print("pass-embedding norms:", {f"p{p + 1}": round(float(np.linalg.norm(tbl[p])), 4)
                                         for p in range(N_PASSES)})
    if PARTS["inject"] is not None:
        print("input-injection gate mean |g|:",
              round(float(np.mean(np.abs(PARTS["inject"].gate.numpy()))), 4))

    # ---- anisotropy by class, per application ----------------------------
    # Does alpha separate cracked from non-cracked MORE on pass 2?
    N_DIAG_BATCHES = min(10, len(test_gen))
    acc_alpha = {}
    lab_all = []
    for bi_ in range(N_DIAG_BATCHES):
        xb, yb = test_gen[bi_]
        recs, _ = trace_forward(xb)
        lab_all.append(yb.astype(int))
        for r in recs:
            if r["alpha"] is not None:
                acc_alpha.setdefault(r["tag"], []).append(r["alpha"].numpy().mean(1))
    lab_all = np.concatenate(lab_all)
    arows = []
    for tag, vals in acc_alpha.items():
        v = np.concatenate(vals)
        m_c = float(v[lab_all == cracked_idx].mean())
        m_n = float(v[lab_all != cracked_idx].mean())
        arows.append(dict(application=tag, alpha_cracked=m_c,
                          alpha_noncracked=m_n, gap=m_c - m_n))
    print("\n" + "=" * 60)
    print(f"  MEAN ANISOTROPY BY CLASS  ({len(lab_all)} test images)")
    print("=" * 60)
    alpha_df = pd.DataFrame(arows)
    print(alpha_df.to_string(index=False))

    # ---- orientation field of the first looped GIPA block, pass by pass --
    first_core = next((b for b in PARTS["core"] if isinstance(b.attn, GIPA)), None)
    if first_core is not None:
        g = first_core.attn.grid
        cell = IMG_SIZE / g
        n_rows = 1 + 2 * N_PASSES
        fig, axes = plt.subplots(n_rows, n_show, figsize=(18, 3.6 * n_rows),
                                 squeeze=False)
        for i in range(n_show):
            recs, _ = trace_forward(bx[i:i + 1])
            per_pass = [r for r in recs if r["tag"].startswith(first_core.name + "/")]
            axes[0, i].imshow(bx[i]); axes[0, i].axis("off")
            axes[0, i].set_title(idx_to_class[int(by[i])], fontsize=9)
            for p, r in enumerate(per_pass):
                u = r["u"][0].numpy(); al = r["alpha"][0].numpy()
                theta = 0.5 * np.arctan2(u[:, 1], u[:, 0])
                an = (al - al.min()) / (al.max() - al.min() + 1e-8)

                ax = axes[1 + p, i]
                ax.imshow(bx[i]); ax.axis("off")
                for idx in range(g * g):
                    r_, c_ = divmod(idx, g)
                    y0, x0_ = (r_ + .5) * cell, (c_ + .5) * cell
                    L = .45 * cell * (.3 + .7 * an[idx])
                    dx_, dy_ = L * np.cos(theta[idx]), L * np.sin(theta[idx])
                    ax.plot([x0_ - dx_, x0_ + dx_], [y0 - dy_, y0 + dy_],
                            lw=1.5, color="lime", alpha=.25 + .75 * an[idx])
                ax.set_title(f"Orientation, pass {p + 1}", fontsize=9)

                ax = axes[1 + N_PASSES + p, i]
                ax.imshow(bx[i])
                ax.imshow(tf.image.resize(al.reshape(g, g)[..., None],
                                          (IMG_SIZE, IMG_SIZE), method="bicubic"
                                          ).numpy().squeeze(), cmap="magma", alpha=.5)
                ax.set_title(rf"Anisotropy $\alpha$, pass {p + 1} "
                             f"(mean {al.mean():.3f})", fontsize=9)
                ax.axis("off")
        plt.suptitle(f"{RUN_NAME} — {first_core.name}: learned geometry, pass by pass",
                     fontsize=13)
        plt.tight_layout(); plt.savefig(f"geometry_{RUN_NAME}.png", dpi=150); plt.show()


# ═══════════════════════════════════════════════════════════════
# 21.  SAVE RESULTS FOR THE ABLATION TABLE
# ═══════════════════════════════════════════════════════════════
ee_mid = ee_df[ee_df["tau"] == 0.90].iloc[0]
res = dict(overall, params=int(model.count_params()), stem=STEM,
           batch_size=BATCH_SIZE, unique_blocks=N_UNIQUE_BLOCKS,
           block_apps_final=EXIT_COST["final"],
           block_apps_exit1=EXIT_COST[EXIT_NAMES[0]],
           pass_embed=PASS_EMBED, per_pass_gates=PER_PASS_GATES,
           input_injection=INPUT_INJECTION, aux_weight=AUX_WEIGHT,
           epochs_run=len(H["loss"]),
           best_threshold=float(thr[bi]), best_f1_cracked=float(f1s[bi]),
           ee090_exited=float(ee_mid["exited_early"]),
           ee090_avg_apps=float(ee_mid["avg_block_apps"]),
           ee090_acc=float(ee_mid["acc"]))
for r in rows:
    res[f"{r['Surface']}_acc"] = r["Accuracy"]
    res[f"{r['Surface']}_auc"] = r["AUC"]

pd.DataFrame([res]).to_csv(f"results_{RUN_NAME}.csv", index=False)
ee_df.to_csv(f"early_exit_{RUN_NAME}.csv", index=False)
if gipa_blocks:
    pd.DataFrame(grep).to_csv(f"gates_{RUN_NAME}.csv", index=False)
    alpha_df.to_csv(f"alpha_{RUN_NAME}.csv", index=False)

print("\n" + "=" * 60)
print(pd.Series(res).to_string())
print("=" * 60)
print(f"\nSaved: results_{RUN_NAME}.csv, early_exit_*.csv, gates_*.csv, "
      f"alpha_*.csv  +  {CKPT_PATH}")
print("""
COMPARISON RUNS (all at BATCH_SIZE = 16):
    untied 6 blocks          -> baseline at the same batch size
    STACK    6x2             -> same params, 12 applications
    SANDWICH 1 + 4x2 + 1     -> same params, 10 applications
    STACK with N_CORE = 3    -> same compute as untied 6, ~40% fewer params
    ATTENTION = "MHSA" in the looped cells -> does looping help GIPA more?
Look at: PER-EXIT table (exit1 vs final), the gate table per pass, and the
anisotropy gap (cracked - non-cracked) on pass 1 vs pass 2.
""")
```

```python
"""
SDNET2018 — LOOPED CrackViT + GIPA — SANDWICH LOOP (1 + 4x2 + 1) — Single Cell
===============================================================================
Recurrent-depth sandwich (Geiping et al. 2025 / SMELT style): the first and
last blocks are untied, the 4 middle blocks are applied twice.

    stem -> cls_pos -> prelude_0 -> [core_0 ... core_3] x 2 -> coda_0 -> head
                                                    |
                                                    +-- after pass 1 -> coda_0 -> head (exit1)

    unique blocks        : 6   (same parameters as the untied 6-block model)
    block applications   : 10 at the final exit, 6 at exit1
                           (exit1 is exactly the depth of the untied model)

Why the sandwich suits GIPA: prelude_0 always reads the raw conv tokens, so
the first orientation estimate is made from low-level texture.  Only the
middle blocks are re-used, on features that are already "about" structure.
The coda is shared by both exits, so the head always sees coda output.

What is added on top of the untied model
  * PassEmbedding    : one learned vector per pass (zero-init), added at the
                       start of each core pass.
  * Per-pass gates   : GIPA gates in the CORE blocks get one row per pass;
                       prelude/coda keep a single row.
  * Exit after pass 1: coda + same head, auxiliary loss weight AUX_WEIGHT.
  * Optional input injection (INPUT_INJECTION, off by default): re-adds the
    prelude output before pass 2.

BATCH_SIZE = 16.  For a fair comparison rerun the untied 6-block model at
batch 16 as well -- batch size changes results on its own.

Paste the whole file into ONE Kaggle cell and run.
"""

# ═══════════════════════════════════════════════════════════════
# 1.  IMPORTS & SETUP
# ═══════════════════════════════════════════════════════════════
import os, math, random, warnings
from functools import partial

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers, models, callbacks
from tensorflow.keras.preprocessing.image import ImageDataGenerator
from tensorflow.keras import mixed_precision

from sklearn.model_selection import train_test_split
from sklearn.utils.class_weight import compute_class_weight
from sklearn.metrics import (
    classification_report, confusion_matrix, ConfusionMatrixDisplay,
    precision_recall_curve, roc_auc_score, f1_score,
)

warnings.filterwarnings("ignore")
print(f"TensorFlow {tf.__version__}  |  Keras {keras.__version__}")
print(f"GPUs: {tf.config.list_physical_devices('GPU')}")

mixed_precision.set_global_policy("mixed_float16")
print(f"Compute dtype: {mixed_precision.global_policy().compute_dtype}")

SEED = 42
random.seed(SEED); np.random.seed(SEED); tf.random.set_seed(SEED)


# ═══════════════════════════════════════════════════════════════
# 2.  CONFIGURATION
# ═══════════════════════════════════════════════════════════════
DATA_ROOT = "/kaggle/input/datasets/aniruddhsharma/structural-defects-network-concrete-crack-images"
WORK_DIR  = "/kaggle/working/split"

ATTENTION  = "GIPA_FULL"   # MHSA | GIPA_S | GIPA_G | GIPA_I | GIPA_GI | GIPA_FULL
STEM       = "conv"        # "conv" (recommended) | "patch"

# ---- looping ------------------------------------------------------------
LOOP_MODE = "sandwich"     # fixed for this cell
N_PRELUDE = 1              # untied blocks before the loop
N_CORE    = 4              # shared blocks inside the loop
N_CODA    = 1              # untied blocks after the loop (runs at every exit)
N_PASSES  = 2              # times the core stack is applied
PASS_EMBED      = True     # learned 256-d vector per pass, zero-init
PER_PASS_GATES  = True     # GIPA gates (lam_c, lam_a, lam_i, gamma, kappa)
                           # untied per pass; attention weights stay shared
INPUT_INJECTION = False    # re-add the loop input on passes >= 2 (zero-init gate)
AUX_WEIGHT      = 0.3      # loss weight on the early exit(s); final exit = 1.0

IMG_SIZE   = 224
BATCH_SIZE = 16            # reduced from 32: 2 passes ~double activation memory
EPOCHS     = 60
LR_INIT    = 3e-4

EMBED_DIM  = 256
NUM_HEADS  = 8
MLP_RATIO  = 2.0
DROPOUT    = 0.1
PATCH_SIZE = 16            # only used when STEM == "patch"
COHERENCE_REG = 0.0        # optional orientation-smoothness penalty (try 1e-3)

EXIT_NAMES = [f"exit{p + 1}" for p in range(N_PASSES - 1)] + ["final"]
# block applications needed to reach each exit (the coda runs at every exit)
EXIT_COST = {name: N_PRELUDE + (p + 1) * N_CORE + N_CODA
             for p, name in enumerate(EXIT_NAMES)}
N_UNIQUE_BLOCKS = N_PRELUDE + N_CORE + N_CODA

RUN_NAME  = (f"loopvit_{LOOP_MODE}_{N_PRELUDE}-{N_CORE}x{N_PASSES}-{N_CODA}"
             f"_{ATTENTION.lower()}_{STEM}_bs{BATCH_SIZE}")
CKPT_PATH = f"best_{RUN_NAME}.weights.h5"
print(f"\n>>> RUN: {RUN_NAME}")
print(f"    unique blocks: {N_UNIQUE_BLOCKS}   block applications per exit: {EXIT_COST}")


# ═══════════════════════════════════════════════════════════════
# 3.  DATA PREPARATION
# ═══════════════════════════════════════════════════════════════
records = []
for surface in ["Decks", "Pavements", "Walls"]:
    for label in ["Cracked", "Non-cracked"]:
        folder = os.path.join(DATA_ROOT, surface, label)
        for fname in os.listdir(folder):
            records.append(dict(path=os.path.join(folder, fname),
                                label=label, surface=surface,
                                fname=f"{surface}_{fname}"))

df = pd.DataFrame(records)
print(f"\nTotal images: {len(df):,}")
print(df.groupby(["surface", "label"]).size().unstack(fill_value=0))

df["strat_key"] = df["surface"] + "_" + df["label"]
train_df, temp_df = train_test_split(df, test_size=0.30, random_state=SEED,
                                     stratify=df["strat_key"])
val_df, test_df = train_test_split(temp_df, test_size=0.50, random_state=SEED,
                                   stratify=temp_df["strat_key"])
print(f"Split -> train {len(train_df):,} | val {len(val_df):,} | test {len(test_df):,}")

for split_name, split_df in [("train", train_df), ("val", val_df), ("test", test_df)]:
    for label in ["Cracked", "Non-cracked"]:
        dst_dir = os.path.join(WORK_DIR, split_name, label)
        os.makedirs(dst_dir, exist_ok=True)
        for _, row in split_df[split_df["label"] == label].iterrows():
            dst = os.path.join(dst_dir, row["fname"])
            if not os.path.exists(dst):
                os.symlink(row["path"], dst)
print("Symlinked directory tree OK\n")


# ═══════════════════════════════════════════════════════════════
# 4.  DATA GENERATORS
# ═══════════════════════════════════════════════════════════════
train_datagen = ImageDataGenerator(
    rescale=1.0/255, rotation_range=20,
    width_shift_range=0.15, height_shift_range=0.15,
    horizontal_flip=True, vertical_flip=True,
    zoom_range=0.15, brightness_range=(0.8, 1.2),
    fill_mode="nearest",
)
val_test_datagen = ImageDataGenerator(rescale=1.0/255)

train_gen = train_datagen.flow_from_directory(
    os.path.join(WORK_DIR, "train"), target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE, class_mode="binary", shuffle=True, seed=SEED)
val_gen = val_test_datagen.flow_from_directory(
    os.path.join(WORK_DIR, "val"), target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE, class_mode="binary", shuffle=False)
test_gen = val_test_datagen.flow_from_directory(
    os.path.join(WORK_DIR, "test"), target_size=(IMG_SIZE, IMG_SIZE),
    batch_size=BATCH_SIZE, class_mode="binary", shuffle=False)

idx_to_class = {v: k for k, v in train_gen.class_indices.items()}
print(f"Class indices: {train_gen.class_indices}")

fig, axes = plt.subplots(2, 5, figsize=(14, 6))
bx, by = next(train_gen)
for i, ax in enumerate(axes.flat):
    ax.imshow(bx[i]); ax.set_title(idx_to_class[int(by[i])], fontsize=9); ax.axis("off")
plt.suptitle("Training Batch (augmented)", fontsize=13); plt.tight_layout(); plt.show()


# ═══════════════════════════════════════════════════════════════
# 5.  CLASS WEIGHTS  +  MULTI-EXIT DATA WRAPPER
# ═══════════════════════════════════════════════════════════════
train_labels = train_df["label"].map(train_gen.class_indices).values
w = compute_class_weight("balanced", classes=np.array([0, 1]), y=train_labels)
class_weights = {0: w[0], 1: w[1]}
print(f"Class weights: {idx_to_class[0]}(0)={w[0]:.3f}  {idx_to_class[1]}(1)={w[1]:.3f}")

# Keras 3 refuses `class_weight=` on a multi-output model, so the same class
# weights are applied as per-sample weights, identically on every exit.
_BaseSeq = getattr(keras.utils, "PyDataset", keras.utils.Sequence)


class MultiExitData(_BaseSeq):
    """Wraps a flow_from_directory iterator: one label copy per exit."""

    def __init__(self, it, cw=None):
        super().__init__()
        self.it, self.cw = it, cw

    def __len__(self):
        return len(self.it)

    def __getitem__(self, i):
        x, y = self.it[i]
        y = y.astype("float32")
        ys = {n: y for n in EXIT_NAMES}
        if self.cw is None:
            return x, ys
        sw = np.where(y == 1, self.cw[1], self.cw[0]).astype("float32")
        return x, ys, {n: sw for n in EXIT_NAMES}

    def on_epoch_end(self):
        self.it.on_epoch_end()


train_seq = MultiExitData(train_gen, class_weights)
val_seq   = MultiExitData(val_gen)
test_seq  = MultiExitData(test_gen)


# ═══════════════════════════════════════════════════════════════
# 6.  TOKENISERS
# ═══════════════════════════════════════════════════════════════

class PatchEmbedding(layers.Layer):
    """Plain ViT tokeniser: non-overlapping patches -> linear projection."""

    def __init__(self, patch_size, embed_dim, **kwargs):
        super().__init__(**kwargs)
        self.patch_size, self.embed_dim = patch_size, embed_dim
        self.proj = layers.Conv2D(embed_dim, patch_size, strides=patch_size,
                                  padding="valid")
        self.norm = layers.LayerNormalization(epsilon=1e-6)

    def call(self, x, training=False):
        x = self.proj(x)
        x = tf.reshape(x, [tf.shape(x)[0], -1, self.embed_dim])
        return self.norm(x)

    def get_config(self):
        c = super().get_config()
        c.update(patch_size=self.patch_size, embed_dim=self.embed_dim)
        return c


class ConvStem(layers.Layer):
    """Convolutional tokeniser: /16 downsample, 224 -> 14x14 tokens.
    Stride-2 convs, not max-pooling: MaxPool keeps the brightest pixel and
    would erase a thin dark crack."""

    def __init__(self, embed_dim, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim = embed_dim

        def blk(c, stride):
            return [layers.Conv2D(c, 3, strides=stride, padding="same",
                                  use_bias=False),
                    layers.BatchNormalization(),
                    layers.Activation("relu")]

        self.stem = keras.Sequential(
            blk(32, 1) + blk(32, 2) +          # 224 -> 112
            blk(64, 1) + blk(64, 2) +          # 112 -> 56
            blk(128, 1) + blk(128, 2) +        # 56  -> 28
            blk(embed_dim, 2),                 # 28  -> 14
            name="conv_stem")
        self.norm = layers.LayerNormalization(epsilon=1e-6)

    def call(self, x, training=False):
        x = self.stem(x, training=training)
        x = tf.reshape(x, [tf.shape(x)[0], -1, self.embed_dim])
        return self.norm(x)

    def get_config(self):
        c = super().get_config(); c.update(embed_dim=self.embed_dim); return c


class CLSPositionEmbedding(layers.Layer):
    """Prepend a learnable [CLS] token, add learnable position embeddings.
    Applied ONCE, before the loop."""

    def __init__(self, num_patches, embed_dim, dropout=0.1, **kwargs):
        super().__init__(**kwargs)
        self.num_patches, self.embed_dim = num_patches, embed_dim
        self.drop = layers.Dropout(dropout)

    def build(self, input_shape):
        self.cls_token = self.add_weight(
            name="cls_token", shape=(1, 1, self.embed_dim),
            initializer="zeros", trainable=True)
        self.pos_embed = self.add_weight(
            name="pos_embed", shape=(1, self.num_patches + 1, self.embed_dim),
            initializer=keras.initializers.TruncatedNormal(stddev=0.02),
            trainable=True)

    def call(self, x, training=False):
        B = tf.shape(x)[0]
        cls = tf.broadcast_to(tf.cast(self.cls_token, x.dtype), [B, 1, self.embed_dim])
        x = tf.concat([cls, x], axis=1) + tf.cast(self.pos_embed, x.dtype)
        return self.drop(x, training=training)

    def get_config(self):
        c = super().get_config()
        c.update(num_patches=self.num_patches, embed_dim=self.embed_dim)
        return c


# ═══════════════════════════════════════════════════════════════
# 7.  LOOP HELPERS  (pass embedding, input injection)
# ═══════════════════════════════════════════════════════════════

class PassEmbedding(layers.Layer):
    """Tells the shared blocks which pass they are on.  One learned vector per
    pass, added to every token at the start of that pass.  Zero-init, so the
    model starts as a plain weight-tied loop and learns to use it."""

    def __init__(self, num_passes, embed_dim, **kwargs):
        super().__init__(**kwargs)
        self.num_passes, self.embed_dim = num_passes, embed_dim

    def build(self, input_shape):
        self.table = self.add_weight(name="pass_table",
                                     shape=(self.num_passes, self.embed_dim),
                                     initializer="zeros", trainable=True)

    def call(self, x, pass_idx=0):
        e = tf.cast(self.table, x.dtype)[pass_idx]
        return x + e[None, None, :]

    def get_config(self):
        c = super().get_config()
        c.update(num_passes=self.num_passes, embed_dim=self.embed_dim)
        return c


class InputInjection(layers.Layer):
    """h <- h + g * x0 on passes >= 2 (Geiping et al. style, additive form).
    Per-channel gate, zero-init: starts identical to plain feedback."""

    def __init__(self, embed_dim, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim = embed_dim

    def build(self, input_shape):
        self.gate = self.add_weight(name="inject_gate", shape=(self.embed_dim,),
                                    initializer="zeros", trainable=True)

    def call(self, h, x0):
        return h + tf.cast(self.gate, h.dtype)[None, None, :] * x0

    def get_config(self):
        c = super().get_config(); c.update(embed_dim=self.embed_dim); return c


# ═══════════════════════════════════════════════════════════════
# 8.  ATTENTION — MHSA BASELINE
# ═══════════════════════════════════════════════════════════════

class MultiHeadSelfAttention(layers.Layer):
    """Standard scaled dot-product MHSA.  `num_passes` / `pass_idx` are
    accepted for interface compatibility and ignored."""

    def __init__(self, embed_dim, num_heads, dropout=0.0, num_passes=1, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim, self.num_heads = embed_dim, num_heads
        self.num_passes = num_passes
        self.head_dim = embed_dim // num_heads
        self.scale = self.head_dim ** -0.5
        self.qkv = layers.Dense(embed_dim * 3, use_bias=False)
        self.proj = layers.Dense(embed_dim)
        self.attn_drop = layers.Dropout(dropout)
        self.proj_drop = layers.Dropout(dropout)
        self.last_attn_weights = None

    def call(self, x, pass_idx=0, training=False):
        B, N = tf.shape(x)[0], tf.shape(x)[1]
        qkv = tf.reshape(self.qkv(x), [B, N, 3, self.num_heads, self.head_dim])
        qkv = tf.transpose(qkv, [2, 0, 3, 1, 4])
        q, k, v = qkv[0], qkv[1], qkv[2]

        attn = tf.nn.softmax(
            tf.cast(tf.matmul(q, k, transpose_b=True), tf.float32) * self.scale, -1)
        self.last_attn_weights = attn
        attn = self.attn_drop(tf.cast(attn, v.dtype), training=training)

        out = tf.transpose(tf.matmul(attn, v), [0, 2, 1, 3])
        out = tf.reshape(out, [B, N, self.embed_dim])
        return self.proj_drop(self.proj(out), training=training)

    def get_config(self):
        c = super().get_config()
        c.update(embed_dim=self.embed_dim, num_heads=self.num_heads,
                 num_passes=self.num_passes)
        return c


# ═══════════════════════════════════════════════════════════════
# 9.  ATTENTION — GIPA  (Geometry-Informed Predictive Attention)
# ═══════════════════════════════════════════════════════════════
#
#   logit_ij = (q_i·k_j)/sqrt(d)    content      [standard MHSA]
#            + lam_G * G_ij         GEOMETRY     orientation coherence +
#                                                displacement alignment,
#                                                anisotropy-gated
#            + lam_I * P_ij         INFORMATION  directional predictive info
#            +         S_ij         STRUCTURE    anisotropic continuity prior
#
# LOOPED VERSION: q/k/v, the structure-tensor head and the predictor W are
# shared across passes.  The scalar gates (lam_c, lam_a, lam_i, gamma, kappa)
# have one row per pass when num_passes > 1, so the model can learn a
# schedule, e.g. strong structure prior on pass 1, more geometry on pass 2.

GATE_INIT = 0.01


class GIPA(layers.Layer):
    """Geometry-Informed Predictive Attention.  Expects N = 1 + g*g."""

    def __init__(self, embed_dim, num_heads, dropout=0.0,
                 use_geometry=True, use_info=True, use_structure=True,
                 coherence_reg=0.0, num_passes=1, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim, self.num_heads = embed_dim, num_heads
        self.head_dim = embed_dim // num_heads
        self.scale = self.head_dim ** -0.5
        self.use_geometry, self.use_info = use_geometry, use_info
        self.use_structure = use_structure
        self.coherence_reg = coherence_reg
        self.num_passes = num_passes

        self.qkv = layers.Dense(embed_dim * 3, use_bias=False)
        self.proj = layers.Dense(embed_dim)
        self.attn_drop = layers.Dropout(dropout)
        self.proj_drop = layers.Dropout(dropout)

        if use_geometry or use_structure:
            self.tensor_head = layers.Dense(
                3, name="structure_tensor", dtype="float32",
                kernel_initializer=keras.initializers.TruncatedNormal(stddev=0.02),
                bias_initializer=lambda shape, dtype=None: tf.constant(
                    [0.5, 0.0, 0.5], dtype=dtype))

        self.last_attn_weights = None
        self.last_orientation = None
        self.last_anisotropy = None

    def build(self, input_shape):
        N = int(input_shape[1]); P = N - 1
        g = int(round(math.sqrt(P)))
        if g * g != P:
            raise ValueError(f"GIPA expects N = 1 + g*g; got N={N}.")
        self.grid, self.num_patches = g, P

        rr, cc = np.meshgrid(np.arange(g), np.arange(g), indexing="ij")
        pos = np.stack([rr.ravel(), cc.ravel()], -1).astype("float32")
        delta = pos[None, :, :] - pos[:, None, :]              # (P,P,2) = p_j - p_i
        d2 = (delta ** 2).sum(-1)
        nrm = np.sqrt(d2) + 1e-8
        dx, dy = delta[..., 0] / nrm, delta[..., 1] / nrm
        disp = np.stack([dx ** 2 - dy ** 2, 2.0 * dx * dy], -1)
        disp[d2 == 0] = 0.0

        self._disp = disp.astype("float32")
        self._dist2n = (d2 / (d2.max() + 1e-8)).astype("float32")
        self._adj = (d2 == 1.0).astype("float32")

        C = keras.initializers.Constant
        G, H = self.num_passes, self.num_heads      # gate table: (passes, heads)
        if self.use_geometry:
            self.lam_c = self.add_weight(name="lam_c", shape=(G, H),
                                         initializer=C(GATE_INIT), trainable=True)
            self.lam_a = self.add_weight(name="lam_a", shape=(G, H),
                                         initializer=C(GATE_INIT), trainable=True)
        if self.use_info:
            eye = np.stack([np.eye(self.head_dim, dtype="float32")
                            for _ in range(self.num_heads)])
            self.W = self.add_weight(
                name="pred_W",
                shape=(self.num_heads, self.head_dim, self.head_dim),
                initializer=lambda shape, dtype=None: tf.constant(eye, tf.float32),
                trainable=True)
            self.lam_i = self.add_weight(name="lam_i", shape=(G, H),
                                         initializer=C(GATE_INIT), trainable=True)
        if self.use_structure:
            self.gamma = self.add_weight(name="gamma", shape=(G, H),
                                         initializer=C(-6.0), trainable=True)
            self.kappa = self.add_weight(name="kappa", shape=(G, H),
                                         initializer="zeros", trainable=True)
        super().build(input_shape)

    def _row(self, var, pass_idx):
        """Gate row for this pass, shaped (1, H, 1, 1), float32."""
        g = pass_idx if self.num_passes > 1 else 0
        return tf.cast(var, tf.float32)[g][None, :, None, None]

    def _orientation(self, x32):
        """u: (B,P,2) unit double-angle (cos2t, sin2t);  alpha: (B,P) in [0,1).
        alpha ~ 0 = isotropic (rough concrete);  ~1 = oriented ridge (crack)."""
        t = tf.cast(self.tensor_head(x32[:, 1:]), tf.float32)
        a, b, c = t[..., 0], t[..., 1], t[..., 2]
        u_raw = tf.stack([a - c, 2.0 * b], -1)
        n = tf.norm(u_raw, axis=-1)
        u = u_raw / (n[..., None] + 1e-6)
        alpha = n / (n + tf.nn.softplus(a + c) + 1e-6)
        return u, alpha

    def _info_bias(self, z):
        """Directional predictive information (negative log residual energy
        of z_j ~ W z_i), row-normalised and floored.  z: (B,H,N,dh)."""
        Wz = tf.einsum('bhnd,hde->bhne', z, tf.cast(self.W, tf.float32))
        zz = tf.reduce_sum(z * z, -1)
        ww = tf.reduce_sum(Wz * Wz, -1)
        cross = tf.matmul(Wz, z, transpose_b=True)
        e = tf.maximum(ww[..., :, None] + zz[..., None, :] - 2.0 * cross, 0.0)
        e = e / (tf.reduce_mean(e, axis=-1, keepdims=True) + 1e-6)
        return -tf.math.log(e + 0.1)

    @staticmethod
    def _pad_cls(bias):
        return tf.pad(bias, [[0, 0], [0, 0], [1, 0], [1, 0]])

    def _coherence_loss(self, u, alpha):
        adj = tf.convert_to_tensor(self._adj)[None]
        c = tf.einsum('bpk,bqk->bpq', u, u)
        wgt = alpha[:, :, None] * alpha[:, None, :]
        num = tf.reduce_sum(adj * wgt * (1.0 - c), [1, 2])
        den = tf.reduce_sum(adj * wgt, [1, 2]) + 1e-6
        return tf.reduce_mean(num / den)

    def call(self, x, pass_idx=0, training=False):
        B, N = tf.shape(x)[0], tf.shape(x)[1]
        qkv = tf.reshape(self.qkv(x), [B, N, 3, self.num_heads, self.head_dim])
        qkv = tf.transpose(qkv, [2, 0, 3, 1, 4])
        q, k, v = qkv[0], qkv[1], qkv[2]

        logits = tf.cast(tf.matmul(q, k, transpose_b=True), tf.float32) * self.scale
        x32 = tf.cast(x, tf.float32)
        u = alpha = align = gate = None

        if self.use_geometry or self.use_structure:
            u, alpha = self._orientation(x32)
            self.last_orientation, self.last_anisotropy = u, alpha
            disp = tf.convert_to_tensor(self._disp)
            align = 0.5 * (tf.einsum('pqk,bpk->bpq', disp, u)
                           + tf.einsum('pqk,bqk->bpq', disp, u))
            gate = tf.sqrt(alpha[:, :, None] * alpha[:, None, :] + 1e-8)

        if self.use_geometry:
            coh = tf.einsum('bpk,bqk->bpq', u, u)
            G = (self._row(self.lam_c, pass_idx) * coh[:, None]
                 + self._row(self.lam_a, pass_idx) * align[:, None]) * gate[:, None]
            logits += self._pad_cls(G)

        if self.use_structure:
            g_h = tf.nn.softplus(self._row(self.gamma, pass_idx))
            k_h = tf.sigmoid(self._row(self.kappa, pass_idx))
            relax = 1.0 - k_h * gate[:, None] * tf.square(align[:, None])
            S = -g_h * tf.convert_to_tensor(self._dist2n)[None, None] * relax
            logits += self._pad_cls(S)

        if self.use_info:
            logits += self._row(self.lam_i, pass_idx) * self._info_bias(
                tf.cast(v, tf.float32))

        attn = tf.nn.softmax(logits, axis=-1)
        self.last_attn_weights = attn
        attn = self.attn_drop(tf.cast(attn, v.dtype), training=training)

        out = tf.transpose(tf.matmul(attn, v), [0, 2, 1, 3])
        out = tf.reshape(out, [B, N, self.embed_dim])
        out = self.proj_drop(self.proj(out), training=training)

        if self.coherence_reg > 0.0 and u is not None:
            self.add_loss(self.coherence_reg * self._coherence_loss(u, alpha))
        return out

    def get_config(self):
        c = super().get_config()
        c.update(embed_dim=self.embed_dim, num_heads=self.num_heads,
                 use_geometry=self.use_geometry, use_info=self.use_info,
                 use_structure=self.use_structure,
                 coherence_reg=self.coherence_reg, num_passes=self.num_passes)
        return c


ATTENTION_REGISTRY = {
    "MHSA":      MultiHeadSelfAttention,
    "GIPA_S":    partial(GIPA, use_geometry=False, use_info=False, use_structure=True),
    "GIPA_G":    partial(GIPA, use_geometry=True,  use_info=False, use_structure=False),
    "GIPA_I":    partial(GIPA, use_geometry=False, use_info=True,  use_structure=False),
    "GIPA_GI":   partial(GIPA, use_geometry=True,  use_info=True,  use_structure=False),
    "GIPA_FULL": partial(GIPA, use_geometry=True,  use_info=True,  use_structure=True,
                         coherence_reg=COHERENCE_REG),
}


# ═══════════════════════════════════════════════════════════════
# 10.  TRANSFORMER BLOCK, HEAD & LOOPED MODEL
# ═══════════════════════════════════════════════════════════════

class TransformerBlock(layers.Layer):
    """Pre-LayerNorm Transformer block; forwards `pass_idx` to the attention."""

    def __init__(self, embed_dim, num_heads, mlp_ratio=2.0, dropout=0.1,
                 attention_cls=None, num_passes=1, **kwargs):
        super().__init__(**kwargs)
        AttnCls = attention_cls or MultiHeadSelfAttention
        self.norm1 = layers.LayerNormalization(epsilon=1e-6)
        self.attn = AttnCls(embed_dim, num_heads, dropout=dropout,
                            num_passes=num_passes)
        self.norm2 = layers.LayerNormalization(epsilon=1e-6)
        self.mlp = keras.Sequential([
            layers.Dense(int(embed_dim * mlp_ratio)),
            layers.Activation("gelu"), layers.Dropout(dropout),
            layers.Dense(embed_dim), layers.Dropout(dropout),
        ])

    def call(self, x, pass_idx=0, training=False):
        x = x + self.attn(self.norm1(x), pass_idx=pass_idx, training=training)
        x = x + self.mlp(self.norm2(x), training=training)
        return x


class ClassifierHead(layers.Layer):
    """final_norm -> [CLS] -> fc -> gelu -> dropout -> sigmoid.
    ONE head shared by every exit, so exits are directly comparable."""

    def __init__(self, embed_dim, **kwargs):
        super().__init__(**kwargs)
        self.embed_dim = embed_dim
        self.norm = layers.LayerNormalization(epsilon=1e-6, name="final_norm")
        self.fc = layers.Dense(embed_dim, name="head_fc")
        self.act = layers.Activation("gelu")
        self.drop = layers.Dropout(0.3)
        self.out = layers.Dense(1, activation="sigmoid", dtype="float32", name="pred")

    def call(self, x, training=False):
        x = self.norm(x)[:, 0]
        x = self.drop(self.act(self.fc(x)), training=training)
        return self.out(x)

    def get_config(self):
        c = super().get_config(); c.update(embed_dim=self.embed_dim); return c


def build_looped_crack_vit(attention_cls, stem="conv",
                           n_prelude=N_PRELUDE, n_core=N_CORE, n_coda=N_CODA,
                           n_passes=N_PASSES, input_shape=(IMG_SIZE, IMG_SIZE, 3),
                           patch_size=PATCH_SIZE, embed_dim=EMBED_DIM,
                           num_heads=NUM_HEADS, mlp_ratio=MLP_RATIO,
                           dropout=DROPOUT):
    """
        stem -> cls_pos -> prelude -> [ core ] x n_passes -> coda -> head
                                            |                  ^
                                            +-- after every pass: coda -> head
                                                (early exits, shared weights)
    """
    def make_block(name, passes):
        return TransformerBlock(embed_dim, num_heads, mlp_ratio, dropout,
                                attention_cls=attention_cls,
                                num_passes=passes, name=name)

    if stem == "conv":
        tokeniser = ConvStem(embed_dim, name="tokeniser")
        grid = input_shape[0] // 16
    else:
        tokeniser = PatchEmbedding(patch_size, embed_dim, name="tokeniser")
        grid = input_shape[0] // patch_size
    cls_pos = CLSPositionEmbedding(grid * grid, embed_dim, dropout, name="cls_pos")

    core_passes = n_passes if PER_PASS_GATES else 1
    prelude = [make_block(f"prelude_{i}", 1) for i in range(n_prelude)]
    core    = [make_block(f"core_{i}", core_passes) for i in range(n_core)]
    coda    = [make_block(f"coda_{i}", 1) for i in range(n_coda)]
    pass_emb = PassEmbedding(n_passes, embed_dim, name="pass_embed") if PASS_EMBED else None
    inject   = InputInjection(embed_dim, name="input_injection") if INPUT_INJECTION else None
    head     = ClassifierHead(embed_dim, name="head")

    inputs = layers.Input(shape=input_shape)
    x = cls_pos(tokeniser(inputs))
    for b in prelude:
        x = b(x, pass_idx=0)

    x0, h, outputs = x, x, {}
    for p in range(n_passes):
        if inject is not None and p > 0:
            h = inject(h, x0)
        if pass_emb is not None:
            h = pass_emb(h, pass_idx=p)
        for b in core:
            h = b(h, pass_idx=p)
        z = h
        for b in coda:
            z = b(z, pass_idx=0)
        outputs[EXIT_NAMES[p]] = head(z)

    model = models.Model(inputs, outputs,
                         name=f"LoopedCrackViT_{ATTENTION}_{LOOP_MODE}")
    parts = dict(tokeniser=tokeniser, cls_pos=cls_pos, prelude=prelude,
                 core=core, coda=coda, pass_emb=pass_emb, inject=inject, head=head)
    return model, parts


model, PARTS = build_looped_crack_vit(ATTENTION_REGISTRY[ATTENTION], stem=STEM)
model.summary()
print(f"\nTotal params: {model.count_params():,}")
print(f"Unique transformer blocks: {N_UNIQUE_BLOCKS}  |  "
      f"block applications: exit1={EXIT_COST[EXIT_NAMES[0]]}, final={EXIT_COST['final']}")


# ═══════════════════════════════════════════════════════════════
# 11.  COMPILE & CALLBACKS
# ═══════════════════════════════════════════════════════════════
def make_metrics():
    return [keras.metrics.BinaryAccuracy(name="acc"),
            keras.metrics.AUC(name="auc"),
            keras.metrics.Precision(name="prec"),
            keras.metrics.Recall(name="rec")]


model.compile(
    optimizer=keras.optimizers.Adam(learning_rate=LR_INIT),
    loss={n: keras.losses.BinaryCrossentropy(label_smoothing=0.05) for n in EXIT_NAMES},
    loss_weights={n: (1.0 if n == "final" else AUX_WEIGHT) for n in EXIT_NAMES},
    metrics={n: make_metrics() for n in EXIT_NAMES},
)

MONITOR = "val_final_auc"
cb_list = [
    callbacks.EarlyStopping(monitor=MONITOR, patience=10, mode="max",
                            restore_best_weights=True, verbose=1),
    callbacks.ReduceLROnPlateau(monitor="val_loss", factor=0.5, patience=4,
                                min_lr=1e-6, verbose=1),
    callbacks.ModelCheckpoint(CKPT_PATH, monitor=MONITOR, mode="max",
                              save_best_only=True, save_weights_only=True,
                              verbose=1),
]


# ═══════════════════════════════════════════════════════════════
# 12.  TRAIN
# ═══════════════════════════════════════════════════════════════
history = model.fit(train_seq, validation_data=val_seq, epochs=EPOCHS,
                    callbacks=cb_list, verbose=1)
H = history.history


# ═══════════════════════════════════════════════════════════════
# 13.  TRAINING CURVES  (final exit solid, early exit dashed)
# ═══════════════════════════════════════════════════════════════
fig, axes = plt.subplots(1, 3, figsize=(18, 5))
axes[0].plot(H["loss"], label="train (total)")
axes[0].plot(H["val_loss"], label="val (total)")
axes[0].set_title("Loss", fontsize=13)
for ax, m, lab in [(axes[1], "acc", "Accuracy"), (axes[2], "auc", "AUC")]:
    for n in EXIT_NAMES:
        ls = "-" if n == "final" else "--"
        if f"{n}_{m}" in H:
            ax.plot(H[f"{n}_{m}"], ls, label=f"train {n}")
            ax.plot(H[f"val_{n}_{m}"], ls, label=f"val {n}")
    ax.set_title(lab, fontsize=13)
for ax in axes:
    ax.set_xlabel("Epoch"); ax.legend(fontsize=8); ax.grid(True, alpha=0.3)
plt.suptitle(RUN_NAME, fontsize=13)
plt.tight_layout(); plt.savefig(f"curves_{RUN_NAME}.png", dpi=150); plt.show()


# ═══════════════════════════════════════════════════════════════
# 14.  TEST EVALUATION  (every exit)
# ═══════════════════════════════════════════════════════════════
model.load_weights(CKPT_PATH)
test_gen.reset()
raw = model.predict(test_seq, verbose=1)
probs = {n: np.asarray(raw[n]).ravel() for n in EXIT_NAMES}
y_true = test_gen.classes
y_prob = probs["final"]
y_pred = (y_prob >= 0.5).astype(int)
target_names = [idx_to_class[i] for i in range(len(idx_to_class))]

print("\n" + "=" * 60)
print(f"  {RUN_NAME} — FINAL EXIT TEST METRICS")
print("=" * 60)
print(classification_report(y_true, y_pred, target_names=target_names, digits=4))

exit_rows = []
for n in EXIT_NAMES:
    pr = probs[n]; pd_ = (pr >= 0.5).astype(int)
    exit_rows.append(dict(exit=n, block_apps=EXIT_COST[n],
                          acc=float((y_true == pd_).mean()),
                          auc=float(roc_auc_score(y_true, pr)),
                          f1_macro=float(f1_score(y_true, pd_, average="macro"))))
exit_df = pd.DataFrame(exit_rows)
print("PER-EXIT (does the second pass help?)")
print(exit_df.to_string(index=False))

overall = dict(run=RUN_NAME, loop_mode=LOOP_MODE, n_prelude=N_PRELUDE,
               n_core=N_CORE, n_coda=N_CODA, n_passes=N_PASSES,
               acc=exit_rows[-1]["acc"], auc=exit_rows[-1]["auc"],
               f1_macro=exit_rows[-1]["f1_macro"])
for r in exit_rows[:-1]:
    overall.update({f"{r['exit']}_acc": r["acc"], f"{r['exit']}_auc": r["auc"],
                    f"{r['exit']}_f1_macro": r["f1_macro"]})
print(f"\nFINAL  ROC-AUC: {overall['auc']:.4f}   F1-macro: {overall['f1_macro']:.4f}")

fig, ax = plt.subplots(figsize=(6, 5))
ConfusionMatrixDisplay(confusion_matrix(y_true, y_pred),
                       display_labels=target_names).plot(ax=ax, cmap="Blues",
                                                         values_format="d")
ax.set_title(f"{RUN_NAME}\nTest Confusion Matrix (final exit)", fontsize=10)
plt.tight_layout(); plt.savefig(f"cm_{RUN_NAME}.png", dpi=150); plt.show()


# ═══════════════════════════════════════════════════════════════
# 15.  PER-SUBSET BREAKDOWN  (Decks / Pavements / Walls)
# ═══════════════════════════════════════════════════════════════
surfaces = np.array([os.path.basename(f).split("_")[0] for f in test_gen.filenames])

print("\n" + "=" * 60)
print("  PER-SUBSET TEST METRICS  (final exit; early-exit AUC alongside)")
print("=" * 60)
rows = []
for subset in ["Decks", "Pavements", "Walls"]:
    mask = surfaces == subset
    if mask.sum() == 0:
        continue
    yt, yp, ypr = y_true[mask], y_pred[mask], y_prob[mask]
    two = len(np.unique(yt)) > 1
    r = dict(Surface=subset, N=int(mask.sum()),
             Accuracy=float((yt == yp).mean()),
             AUC=roc_auc_score(yt, ypr) if two else np.nan,
             F1_macro=f1_score(yt, yp, average="macro"))
    for n in EXIT_NAMES[:-1]:
        r[f"AUC_{n}"] = roc_auc_score(yt, probs[n][mask]) if two else np.nan
    rows.append(r)
    print(f"\n--- {subset} ({mask.sum():,} images) ---")
    print(classification_report(yt, yp, target_names=target_names,
                                digits=4, zero_division=0))
subset_df = pd.DataFrame(rows)
print(subset_df.to_string(index=False))


# ═══════════════════════════════════════════════════════════════
# 16.  PRECISION-RECALL CURVE + THRESHOLD TUNING  (final exit)
# ═══════════════════════════════════════════════════════════════
cracked_idx = train_gen.class_indices["Cracked"]
y_true_c = (y_true == cracked_idx).astype(int)
y_prob_c = y_prob if cracked_idx == 1 else 1.0 - y_prob

prec, rec, thr = precision_recall_curve(y_true_c, y_prob_c)
f1s = 2 * prec[:-1] * rec[:-1] / (prec[:-1] + rec[:-1] + 1e-8)
bi = int(np.argmax(f1s))
print(f"\nOptimal threshold (Cracked-positive): {thr[bi]:.3f}")
print(f"  -> F1 {f1s[bi]:.4f} | Prec {prec[bi]:.4f} | Rec {rec[bi]:.4f}")

fig, ax = plt.subplots(figsize=(8, 5))
ax.plot(rec, prec, lw=2)
ax.scatter(rec[bi], prec[bi], s=120, c="red", zorder=5,
           label=f"Best F1={f1s[bi]:.3f} @ t={thr[bi]:.3f}")
ax.set_xlabel("Recall (Cracked)"); ax.set_ylabel("Precision (Cracked)")
ax.set_title(f"{RUN_NAME}\nPR Curve, Cracked class (final exit)", fontsize=10)
ax.legend(); ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(f"pr_{RUN_NAME}.png", dpi=150); plt.show()


# ═══════════════════════════════════════════════════════════════
# 17.  CONFIDENCE-BASED EARLY EXIT  (no extra training)
# ═══════════════════════════════════════════════════════════════
# If exit 1 is confident (max(p, 1-p) >= tau) stop there, else take the
# final exit.  Label smoothing 0.05 caps confidence near 0.975, so useful
# tau values sit below that.
p1, pf = probs[EXIT_NAMES[0]], probs["final"]
conf1 = np.maximum(p1, 1.0 - p1)
c1, cF = EXIT_COST[EXIT_NAMES[0]], EXIT_COST["final"]
ee_rows = []
for tau in [None, 0.97, 0.95, 0.925, 0.90, 0.85, 0.80, 0.0]:
    stop = np.zeros_like(conf1, bool) if tau is None else conf1 >= tau
    pm = np.where(stop, p1, pf); pdm = (pm >= 0.5).astype(int)
    frac = float(stop.mean())
    ee_rows.append(dict(
        tau="final only" if tau is None else ("exit1 only" if tau == 0.0 else tau),
        exited_early=frac,
        avg_block_apps=frac * c1 + (1 - frac) * cF,
        acc=float((y_true == pdm).mean()),
        auc=float(roc_auc_score(y_true, pm)),
        f1_macro=float(f1_score(y_true, pdm, average="macro"))))
ee_df = pd.DataFrame(ee_rows)
print("\n" + "=" * 60)
print("  EARLY EXIT: accuracy vs average compute")
print("=" * 60)
print(ee_df.to_string(index=False))

fig, ax = plt.subplots(figsize=(7, 4.5))
ax.plot(ee_df["avg_block_apps"], ee_df["acc"], "o-")
for _, r in ee_df.iterrows():
    ax.annotate(str(r["tau"]), (r["avg_block_apps"], r["acc"]),
                fontsize=8, xytext=(4, 4), textcoords="offset points")
ax.set_xlabel("Average block applications per image")
ax.set_ylabel("Test accuracy")
ax.set_title(f"{RUN_NAME}\nEarly-exit trade-off", fontsize=10)
ax.grid(True, alpha=0.3)
plt.tight_layout(); plt.savefig(f"early_exit_{RUN_NAME}.png", dpi=150); plt.show()


# ═══════════════════════════════════════════════════════════════
# 18.  TRACED FORWARD PASS  (records EVERY block application)
# ═══════════════════════════════════════════════════════════════
# Shared layers overwrite last_attn_weights / last_orientation on every call,
# so the functional model only remembers the last pass.  This replays the
# final-exit path eagerly and snapshots each application as it happens.

def trace_forward(img):
    P = PARTS
    x = P["cls_pos"](P["tokeniser"](img, training=False), training=False)
    rec = []

    def snap(blk, tag):
        a = blk.attn
        rec.append(dict(tag=tag, attn=a.last_attn_weights,
                        u=getattr(a, "last_orientation", None),
                        alpha=getattr(a, "last_anisotropy", None),
                        is_gipa=isinstance(a, GIPA)))

    for b in P["prelude"]:
        x = b(x, pass_idx=0, training=False); snap(b, f"{b.name}")
    x0, h = x, x
    for p in range(N_PASSES):
        if P["inject"] is not None and p > 0:
            h = P["inject"](h, x0)
        if P["pass_emb"] is not None:
            h = P["pass_emb"](h, pass_idx=p)
        for b in P["core"]:
            h = b(h, pass_idx=p, training=False); snap(b, f"{b.name}/p{p + 1}")
    z = h
    for b in P["coda"]:
        z = b(z, pass_idx=0, training=False); snap(b, f"{b.name}")
    prob = P["head"](z, training=False)
    return rec, prob


# ═══════════════════════════════════════════════════════════════
# 19.  ATTENTION ROLLOUT  (all block applications on the final path)
# ═══════════════════════════════════════════════════════════════

def attention_rollout(recs, head_fusion="mean"):
    maps = []
    for r in recs:
        a = r["attn"]
        a = tf.reduce_mean(a, 1) if head_fusion == "mean" else tf.reduce_max(a, 1)
        maps.append(tf.cast(a, tf.float32))
    N = maps[0].shape[-1]
    roll = tf.eye(N, batch_shape=[tf.shape(maps[0])[0]])
    for a in maps:
        roll = tf.matmul(roll, 0.5 * a + 0.5 * tf.eye(N))
    cls_attn = roll[0, 0, 1:].numpy()
    g = int(np.sqrt(len(cls_attn)))
    hm = cls_attn.reshape(g, g)
    return (hm - hm.min()) / (hm.max() - hm.min() + 1e-8)


test_gen.reset()
bx, by = test_gen[0]
n_show = min(5, len(bx))
fig, axes = plt.subplots(2, n_show, figsize=(18, 7), squeeze=False)
for i in range(n_show):
    img = bx[i:i + 1]
    recs, prob = trace_forward(img)
    p = float(prob.numpy()[0, 0])
    hm = attention_rollout(recs)
    hm_up = tf.image.resize(hm[..., None], (IMG_SIZE, IMG_SIZE),
                            method="bicubic").numpy().squeeze()
    axes[0, i].imshow(bx[i]); axes[0, i].axis("off")
    axes[0, i].set_title(f"True: {idx_to_class[int(by[i])]}\n"
                         f"Pred: {idx_to_class[int(p >= 0.5)]} ({p:.2f})", fontsize=9)
    axes[1, i].imshow(bx[i]); axes[1, i].imshow(hm_up, cmap="jet", alpha=0.4)
    axes[1, i].set_title(f"Rollout over {len(recs)} applications", fontsize=9)
    axes[1, i].axis("off")
plt.suptitle(f"{RUN_NAME} — where the model attends", fontsize=13)
plt.tight_layout(); plt.savefig(f"rollout_{RUN_NAME}.png", dpi=150); plt.show()


# ═══════════════════════════════════════════════════════════════
# 20.  GIPA-ONLY DIAGNOSTICS  (per-pass gates, geometry per pass)
# ═══════════════════════════════════════════════════════════════
gipa_blocks = [b for b in PARTS["prelude"] + PARTS["core"] + PARTS["coda"]
               if isinstance(b.attn, GIPA)]

if gipa_blocks:
    sp = lambda v: np.log1p(np.exp(v))
    sg = lambda v: 1 / (1 + np.exp(-v))

    # ---- learned gate values, one row per (block, pass) ------------------
    grep = []
    for b in gipa_blocks:
        a = b.attn
        for g in range(a.num_passes):
            r = {"block": b.name, "pass": g + 1 if a.num_passes > 1 else "shared"}
            if a.use_geometry:
                r["lam_c"] = float(np.mean(np.abs(a.lam_c.numpy()[g])))
                r["lam_a"] = float(np.mean(np.abs(a.lam_a.numpy()[g])))
            if a.use_info:
                r["lam_i"] = float(np.mean(np.abs(a.lam_i.numpy()[g])))
            if a.use_structure:
                r["gamma_sp"] = float(np.mean(sp(a.gamma.numpy()[g])))
                r["kappa_sig"] = float(np.mean(sg(a.kappa.numpy()[g])))
            grep.append(r)
    print("\n" + "=" * 60)
    print("  LEARNED GATE VALUES PER PASS  (init 0.01; growth => term is used)")
    print("=" * 60)
    print(pd.DataFrame(grep).to_string(index=False))
    if PARTS["pass_emb"] is not None:
        tbl = PARTS["pass_emb"].table.numpy()
        print("pass-embedding norms:", {f"p{p + 1}": round(float(np.linalg.norm(tbl[p])), 4)
                                         for p in range(N_PASSES)})
    if PARTS["inject"] is not None:
        print("input-injection gate mean |g|:",
              round(float(np.mean(np.abs(PARTS["inject"].gate.numpy()))), 4))

    # ---- anisotropy by class, per application ----------------------------
    # Does alpha separate cracked from non-cracked MORE on pass 2?
    N_DIAG_BATCHES = min(10, len(test_gen))
    acc_alpha = {}
    lab_all = []
    for bi_ in range(N_DIAG_BATCHES):
        xb, yb = test_gen[bi_]
        recs, _ = trace_forward(xb)
        lab_all.append(yb.astype(int))
        for r in recs:
            if r["alpha"] is not None:
                acc_alpha.setdefault(r["tag"], []).append(r["alpha"].numpy().mean(1))
    lab_all = np.concatenate(lab_all)
    arows = []
    for tag, vals in acc_alpha.items():
        v = np.concatenate(vals)
        m_c = float(v[lab_all == cracked_idx].mean())
        m_n = float(v[lab_all != cracked_idx].mean())
        arows.append(dict(application=tag, alpha_cracked=m_c,
                          alpha_noncracked=m_n, gap=m_c - m_n))
    print("\n" + "=" * 60)
    print(f"  MEAN ANISOTROPY BY CLASS  ({len(lab_all)} test images)")
    print("=" * 60)
    alpha_df = pd.DataFrame(arows)
    print(alpha_df.to_string(index=False))

    # ---- orientation field of the first looped GIPA block, pass by pass --
    first_core = next((b for b in PARTS["core"] if isinstance(b.attn, GIPA)), None)
    if first_core is not None:
        g = first_core.attn.grid
        cell = IMG_SIZE / g
        n_rows = 1 + 2 * N_PASSES
        fig, axes = plt.subplots(n_rows, n_show, figsize=(18, 3.6 * n_rows),
                                 squeeze=False)
        for i in range(n_show):
            recs, _ = trace_forward(bx[i:i + 1])
            per_pass = [r for r in recs if r["tag"].startswith(first_core.name + "/")]
            axes[0, i].imshow(bx[i]); axes[0, i].axis("off")
            axes[0, i].set_title(idx_to_class[int(by[i])], fontsize=9)
            for p, r in enumerate(per_pass):
                u = r["u"][0].numpy(); al = r["alpha"][0].numpy()
                theta = 0.5 * np.arctan2(u[:, 1], u[:, 0])
                an = (al - al.min()) / (al.max() - al.min() + 1e-8)

                ax = axes[1 + p, i]
                ax.imshow(bx[i]); ax.axis("off")
                for idx in range(g * g):
                    r_, c_ = divmod(idx, g)
                    y0, x0_ = (r_ + .5) * cell, (c_ + .5) * cell
                    L = .45 * cell * (.3 + .7 * an[idx])
                    dx_, dy_ = L * np.cos(theta[idx]), L * np.sin(theta[idx])
                    ax.plot([x0_ - dx_, x0_ + dx_], [y0 - dy_, y0 + dy_],
                            lw=1.5, color="lime", alpha=.25 + .75 * an[idx])
                ax.set_title(f"Orientation, pass {p + 1}", fontsize=9)

                ax = axes[1 + N_PASSES + p, i]
                ax.imshow(bx[i])
                ax.imshow(tf.image.resize(al.reshape(g, g)[..., None],
                                          (IMG_SIZE, IMG_SIZE), method="bicubic"
                                          ).numpy().squeeze(), cmap="magma", alpha=.5)
                ax.set_title(rf"Anisotropy $\alpha$, pass {p + 1} "
                             f"(mean {al.mean():.3f})", fontsize=9)
                ax.axis("off")
        plt.suptitle(f"{RUN_NAME} — {first_core.name}: learned geometry, pass by pass",
                     fontsize=13)
        plt.tight_layout(); plt.savefig(f"geometry_{RUN_NAME}.png", dpi=150); plt.show()


# ═══════════════════════════════════════════════════════════════
# 21.  SAVE RESULTS FOR THE ABLATION TABLE
# ═══════════════════════════════════════════════════════════════
ee_mid = ee_df[ee_df["tau"] == 0.90].iloc[0]
res = dict(overall, params=int(model.count_params()), stem=STEM,
           batch_size=BATCH_SIZE, unique_blocks=N_UNIQUE_BLOCKS,
           block_apps_final=EXIT_COST["final"],
           block_apps_exit1=EXIT_COST[EXIT_NAMES[0]],
           pass_embed=PASS_EMBED, per_pass_gates=PER_PASS_GATES,
           input_injection=INPUT_INJECTION, aux_weight=AUX_WEIGHT,
           epochs_run=len(H["loss"]),
           best_threshold=float(thr[bi]), best_f1_cracked=float(f1s[bi]),
           ee090_exited=float(ee_mid["exited_early"]),
           ee090_avg_apps=float(ee_mid["avg_block_apps"]),
           ee090_acc=float(ee_mid["acc"]))
for r in rows:
    res[f"{r['Surface']}_acc"] = r["Accuracy"]
    res[f"{r['Surface']}_auc"] = r["AUC"]

pd.DataFrame([res]).to_csv(f"results_{RUN_NAME}.csv", index=False)
ee_df.to_csv(f"early_exit_{RUN_NAME}.csv", index=False)
if gipa_blocks:
    pd.DataFrame(grep).to_csv(f"gates_{RUN_NAME}.csv", index=False)
    alpha_df.to_csv(f"alpha_{RUN_NAME}.csv", index=False)

print("\n" + "=" * 60)
print(pd.Series(res).to_string())
print("=" * 60)
print(f"\nSaved: results_{RUN_NAME}.csv, early_exit_*.csv, gates_*.csv, "
      f"alpha_*.csv  +  {CKPT_PATH}")
print("""
COMPARISON RUNS (all at BATCH_SIZE = 16):
    untied 6 blocks          -> baseline at the same batch size
    STACK    6x2             -> same params, 12 applications
    SANDWICH 1 + 4x2 + 1     -> same params, 10 applications
    STACK with N_CORE = 3    -> same compute as untied 6, ~40% fewer params
    ATTENTION = "MHSA" in the looped cells -> does looping help GIPA more?
Look at: PER-EXIT table (exit1 vs final), the gate table per pass, and the
anisotropy gap (cracked - non-cracked) on pass 1 vs pass 2.
""")
```
